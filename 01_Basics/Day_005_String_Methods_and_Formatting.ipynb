# **Day 5: String Methods and Formatting**

A **method** is a function that belongs to a value. For strings, methods help us change case, search, split, replace and much more.

We use a method with a dot: `text.upper()`.

**Important:** String methods do **not** change the original string. They give back a **new** string.

## **Changing Case and Removing Spaces**

```python
s = "  howrah bridge  "

print(s.upper())        # All capital letters
print(s.lower())        # All small letters
print(s.title())        # First letter of each word capital
print(s.strip())        # Remove spaces from both sides
print(s.strip().capitalize())   # Only the first letter capital

print(s)                # The original string is unchanged
```

## **Finding, Counting and Replacing**
- `find()`: Gives the position of some text. Gives `-1` if not found.
- `count()`: Counts how many times some text appears.
- `replace()`: Replaces some text with other text.

```python
line = "kolkata is the city of joy. kolkata loves football."

print(line.find("city"))
print(line.find("mumbai"))
print(line.count("kolkata"))
print(line.replace("football", "cricket"))
```

## **Split and Join**
- `split()`: Breaks a string into a **list** of smaller strings.
- `join()`: Joins a list of strings into one string.

These two are used a lot when reading data files.

```python
stations = "Howrah,Santragachi,Mecheda,Panskura,Kharagpur"

station_list = stations.split(",")
print(station_list)
print(len(station_list), "stations")

print(" -> ".join(station_list))
```

## **Example from Remote Sensing: Reading a Sentinel-2 File Name**
Sentinel-2 file names are also long, and their parts are separated by `_`.

`S2A_MSIL2A_20240115T045141_N0510_R076_T45QXF_20240115T083015`

We can use `split("_")` to break it into parts.

```python
file_name = "S2A_MSIL2A_20240115T045141_N0510_R076_T45QXF_20240115T083015"

parts = file_name.split("_")
print(parts)

satellite = parts[0]
product = parts[1]
date_time = parts[2]
tile = parts[5]

print("Satellite:", satellite)
print("Product level:", product[3:])
print("Date:", date_time[6:8] + "-" + date_time[4:6] + "-" + date_time[0:4])
print("Tile:", tile[1:])
```

- `file_name.split("_")`: Cuts the name wherever there is an underscore.
- `parts[2]`: The third part, `20240115T045141`, which contains the date and time.
- `tile[1:]`: Removes the first letter `T` from `T45QXF` to get the tile number `45QXF`.

## **Checking Methods (They Return True or False)**

```python
print("700001".isdigit())             # Only digits?
print("Kolkata".isalpha())            # Only letters?
print("WB02".isalnum())               # Letters and digits only?
print("Siliguri".startswith("Sili"))  # Starts with?
print("band4.tif".endswith(".tif"))   # Ends with?
```

## **String Formatting**
Formatting means putting values **inside** a string in a clean way. There are three ways, but the **f-string** is the easiest and most used.

To make an f-string, write `f` before the quotes and put variables inside `{ }`.

```python
name = "Rahul"
city = "Kharagpur"
age = 20

# 1. Old % style
print("%s lives in %s and is %d years old" % (name, city, age))

# 2. format() method
print("{} lives in {} and is {} years old".format(name, city, age))

# 3. f-string (recommended)
print(f"{name} lives in {city} and is {age} years old")
```

## **Useful f-string Tricks**

```python
price = 1234.5678

print(f"{price:.2f}")        # 2 decimal places
print(f"{price:,.2f}")       # With comma
print(f"{12:05d}")           # Fill with zeros
print(f"{0.4567:.1%}")       # As percentage
print(f"{2 * 21}")           # Calculation inside

x = 42
print(f"{x = }")             # Shows the name and the value (useful for checking)
```

```python
# Alignment: < left, > right, ^ centre
print(f"{'Item':<12}{'Price':>8}")
print(f"{'Jhalmuri':<12}{20:>8}")
print(f"{'Kathi Roll':<12}{80:>8}")
print(f"{'Mishti Doi':<12}{45:>8}")
```

## **Example: Fish Market Bill**

```python
ilish_kg, ilish_rate = 1.2, 1600
rui_kg, rui_rate = 1.5, 280
prawn_kg, prawn_rate = 0.5, 700

ilish = ilish_kg * ilish_rate
rui = rui_kg * rui_rate
prawn = prawn_kg * prawn_rate
total = ilish + rui + prawn

print("-" * 34)
print(f"{'Fish':<10}{'Kg':>6}{'Rate':>8}{'Amount':>10}")
print("-" * 34)
print(f"{'Ilish':<10}{ilish_kg:>6}{ilish_rate:>8}{ilish:>10.2f}")
print(f"{'Rui':<10}{rui_kg:>6}{rui_rate:>8}{rui:>10.2f}")
print(f"{'Prawn':<10}{prawn_kg:>6}{prawn_rate:>8}{prawn:>10.2f}")
print("-" * 34)
print(f"{'Total':<24}{total:>10.2f}")
```

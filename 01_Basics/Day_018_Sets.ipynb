# **Day 18: Sets**

A **set** is a collection of **unique** items. It uses curly brackets `{ }`.

**Features of a set:**
- **No duplicates:** Every item appears only once.
- **No order:** Items do not have positions, so there is no indexing.
- **Changeable:** We can add and remove items.

Sets are very useful for **removing duplicates** and for finding **common** or **different** items between two groups.

```python
fish = {"ilish", "rui", "katla", "ilish", "pabda", "rui"}
print(fish)          # Duplicates are removed automatically
print(len(fish))

# print(fish[0])     # This will raise a TypeError (sets have no index)
```

```python
# An empty set must be made with set(), because {} makes an empty dictionary
x = {}
y = set()
print(type(x), type(y))
```

## **Adding and Removing Items**

```python
fish.add("tangra")
fish.add("rui")          # Already there, nothing happens
print(fish)

fish.remove("katla")     # Gives an error if the item is missing
fish.discard("shark")    # No error if the item is missing
print(fish)
```

## **Set Operations**
Two friends visited these Durga Puja pandals:

```python
riya = {"Bagbazar", "Kumartuli", "Santosh Mitra Square", "Deshapriya Park", "Ekdalia"}
arjun = {"Ekdalia", "Suruchi Sangha", "Bagbazar", "Tridhara", "Chetla Agrani"}

print("Both visited   :", riya & arjun)     # Intersection
print("Anyone visited :", riya | arjun)     # Union
print("Only Riya      :", riya - arjun)     # Difference
print("Not common     :", riya ^ arjun)     # Symmetric difference
```

| Operation | Symbol | Method | Meaning |
|---|---|---|---|
| Union | `\|` | `union()` | All items from both |
| Intersection | `&` | `intersection()` | Only common items |
| Difference | `-` | `difference()` | Items in the first but not in the second |
| Symmetric difference | `^` | `symmetric_difference()` | Items in only one of them |

## **Example from Remote Sensing: Which Tiles Cover Both Districts?**
Sentinel-2 divides the Earth into squares called **tiles**, each with a name like `45QXF`. To download images for an area, we need to know which tiles cover it.

Suppose we know the tiles for two districts. Let us find the tiles we need for both, and the tiles they share. (Tile names here are only examples.)

```python
kolkata_tiles = {"45QXE", "45QXF", "45QYF"}
howrah_tiles = {"45QXF", "45QWF", "45QXE"}

all_tiles = kolkata_tiles | howrah_tiles
shared = kolkata_tiles & howrah_tiles
only_kolkata = kolkata_tiles - howrah_tiles

print("Tiles to download:", sorted(all_tiles))
print("Shared tiles     :", sorted(shared))
print("Only for Kolkata :", sorted(only_kolkata))
```

Using a set, each tile is downloaded **only once**, even if it covers both districts.

## **Subset and Superset**

```python
north_bengal = {"Darjeeling", "Kalimpong", "Jalpaiguri", "Alipurduar", "Cooch Behar",
                "Uttar Dinajpur", "Dakshin Dinajpur", "Malda"}
hills = {"Darjeeling", "Kalimpong"}

print(hills.issubset(north_bengal))          # Is every hill district in North Bengal?
print(north_bengal.issuperset(hills))
print(hills.isdisjoint({"Kolkata", "Howrah"}))   # No common items?
```

## **Sets Are Very Fast for Searching**
Checking `x in my_set` is much faster than checking `x in my_list`, especially for big collections.

```python
import time

big_list = list(range(1_000_000))
big_set = set(big_list)

t = time.time()
for _ in range(200):
    999_999 in big_list
print("List:", round(time.time() - t, 4), "seconds")

t = time.time()
for _ in range(200):
    999_999 in big_set
print("Set :", round(time.time() - t, 6), "seconds")
```

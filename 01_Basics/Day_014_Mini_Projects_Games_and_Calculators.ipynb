## **Project 1: Number Guessing Game**
The computer picks a secret number from 1 to 50. The player guesses, and the computer says "higher" or "lower" until the player finds it.

```python
import random

def play_game():
    secret = random.randint(1, 50)
    tries = 0
    while True:
        guess = int(input("Guess (1-50): "))
        tries += 1
        if guess < secret:
            print("Higher")
        elif guess > secret:
            print("Lower")
        else:
            print(f"Correct! You took {tries} tries")
            break

# play_game()       # Remove the # to play
```

Below, the computer plays by itself using a smart method: it always guesses the **middle** of the possible range. This method is called **binary search**.

```python
random.seed(7)                # Fixed seed so we get the same number every time
secret = random.randint(1, 50)

low, high = 1, 50
tries = 0
while True:
    guess = (low + high) // 2
    tries += 1
    print(f"Try {tries}: {guess}", end=" -> ")
    if guess < secret:
        print("Higher")
        low = guess + 1
    elif guess > secret:
        print("Lower")
        high = guess - 1
    else:
        print("Correct!")
        break
```

Binary search always finds the number from 1 to 50 in at most 6 tries, because every guess cuts the range in half.

## **Project 2: Kolkata Metro Fare Calculator**
The stations of the Blue Line are kept in order in a list. We assume about 1.2 km between two stations. (Fares here are only approximate.)

```python
stations = ["Dakshineswar", "Baranagar", "Noapara", "Dum Dum", "Belgachia",
            "Shyambazar", "Shobhabazar", "Girish Park", "MG Road", "Central",
            "Chandni Chowk", "Esplanade", "Park Street", "Maidan", "Rabindra Sadan",
            "Netaji Bhavan", "Jatin Das Park", "Kalighat", "Rabindra Sarobar",
            "Mahanayak Uttam Kumar", "Netaji", "Masterda Surya Sen", "Gitanjali",
            "Kavi Nazrul", "Shahid Khudiram", "Kavi Subhash"]
gap_km = 1.2

from_station = "Dum Dum"
to_station = "Kalighat"

a = stations.index(from_station)
b = stations.index(to_station)
km = abs(b - a) * gap_km

if km <= 2:
    fare = 5
elif km <= 5:
    fare = 10
elif km <= 10:
    fare = 15
elif km <= 20:
    fare = 20
else:
    fare = 25

print(f"{from_station} -> {to_station}")
print(f"Stations: {abs(b - a)}, Distance: {km:.1f} km, Fare: Rs {fare}")
```

- `stations.index(from_station)`: Finds the position of the station in the list.
- `abs(b - a)`: The number of stations between them. `abs()` makes it positive, so the direction does not matter.

## **Project 3: Satellite Pixel Area Calculator**
A GIS student often needs to convert "number of pixels" into real ground area. Let us build a small calculator for different satellites.

```python
satellite = "Sentinel-2"
pixel_count = 12500           # Number of pixels in a lake

if satellite == "Landsat 8":
    pixel_size = 30
elif satellite == "Sentinel-2":
    pixel_size = 10
elif satellite == "MODIS":
    pixel_size = 250
else:
    pixel_size = None

if pixel_size is None:
    print("Unknown satellite")
else:
    area_m2 = pixel_count * pixel_size ** 2
    print(f"Satellite       : {satellite}")
    print(f"Pixel size      : {pixel_size} m")
    print(f"Pixels          : {pixel_count:,}")
    print(f"Area            : {area_m2:,} square metre")
    print(f"Area            : {area_m2 / 10_000:,.2f} hectare")
    print(f"Area            : {area_m2 / 1_000_000:,.3f} square km")
```

## **Project 4: Sweet Shop Bill**

```python
menu = {"rosogolla": 15, "sandesh": 20, "mishti doi": 45, "langcha": 25}
order = [("rosogolla", 6), ("mishti doi", 2), ("langcha", 4)]

total = 0
print(f"{'Item':<12}{'Qty':>4}{'Amount':>8}")
for item, qty in order:
    amount = menu[item] * qty
    total += amount
    print(f"{item:<12}{qty:>4}{amount:>8}")
print("-" * 24)

if total > 200:
    discount = total * 0.05
    print(f"{'Discount 5%':<16}{-discount:>8.1f}")
    total -= discount
print(f"{'Total':<16}{total:>8.1f}")
```

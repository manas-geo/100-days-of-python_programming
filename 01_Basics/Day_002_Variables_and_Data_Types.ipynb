# **Day 2: Variables and Data Types**

## **What is a Variable?**
A **variable** is like a **labelled box** where we keep a value. The label is the variable name, and the thing inside the box is the value.

For example, `city = "Kolkata"` means: make a box, name it `city`, and put `"Kolkata"` inside it.

In Python we **do not need to tell the type** of the value. Python understands it automatically. This is called **dynamic typing**.

```python
# Creating variables
city = "Kolkata"
population = 4496694      # Census 2011
area_km2 = 206.08
is_capital = True

print(city)
print(population)
print(area_km2)
print(is_capital)
```

## **Checking the Type with type()**
The `type()` function tells us what kind of value is stored in a variable.

```python
print(type(city))
print(type(population))
print(type(area_km2))
print(type(is_capital))
```

## **Basic Data Types in Python**

| Data Type | Meaning | Example |
|---|---|---|
| `int` | Whole number | `721101` |
| `float` | Number with a decimal point | `22.57` |
| `str` | Text (string) | `"Siliguri"` |
| `bool` | True or False | `True` |
| `complex` | Complex number | `3+4j` |
| `NoneType` | Nothing / no value | `None` |

```python
pin = 734001              # int
latitude = 26.7271        # float
name = "Siliguri"         # str
has_airport = True        # bool
z = 3 + 4j                # complex
nothing = None            # NoneType

print(pin, type(pin))
print(latitude, type(latitude))
print(name, type(name))
print(has_airport, type(has_airport))
print(z, type(z))
print(nothing, type(nothing))
```

## **Example from Remote Sensing**
A satellite image is described by some simple information. Let us store it in variables.

```python
satellite = "Landsat 8"
sensor = "OLI"
resolution_m = 30           # Size of one pixel on the ground (in metre)
number_of_bands = 11
cloud_cover = 12.5          # In percent
is_free_data = True

print("Satellite:", satellite)
print("Sensor:", sensor)
print("Pixel size:", resolution_m, "m")
print("Bands:", number_of_bands)
print("Cloud cover:", cloud_cover, "%")
print("Free to download:", is_free_data)
```

- **Resolution** means how much ground area one pixel covers. A 30 m resolution means one pixel = 30 m x 30 m on the ground.
- **Bands** are the different "colours" (wavelengths of light) that the satellite records.

## **Dynamic Typing**
The same variable can hold a different type of value later.

```python
x = 10
print(x, type(x))

x = "ten"
print(x, type(x))

x = 10.0
print(x, type(x))
```

## **Multiple Assignment**
We can create many variables in one line.

```python
# Different values to different variables
a, b, c = "Kolkata", "Howrah", "Hooghly"
print(a, b, c)

# Same value to many variables
x = y = z = 0
print(x, y, z)
```

## **Swapping Two Variables**
In many languages, we need a third variable to swap two values. In Python, it is just one line.

```python
morning_train = "Howrah"
evening_train = "Kharagpur"

morning_train, evening_train = evening_train, morning_train
print(morning_train, evening_train)
```

## **Rules for Naming Variables**
1. A name can start with a **letter** or **underscore** `_`.
2. A name **cannot start with a number**.
3. Only letters, numbers and `_` are allowed (no spaces).
4. Names are **case sensitive** (`city` and `City` are different).
5. We cannot use Python **keywords** like `if`, `for`, `class` as names.

In Python, the common style is **snake_case**, which means small letters with underscores (for example `district_name`).

```python
district_name = "Paschim Medinipur"   # Good style
DistrictName = "Purba Medinipur"      # Works, but not the usual Python style
_temp = 42                            # Works

print(district_name, DistrictName, _temp)

# 2nd_city = "Howrah"                 # This will raise a SyntaxError (starts with a number)
```

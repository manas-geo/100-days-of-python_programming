# **Day 16: List Methods**

Lists come with many built-in **methods** (functions that belong to the list). They help us add, remove, search and sort items.

```python
# All the list methods
print([m for m in dir(list) if not m.startswith("_")])
```

## **Adding Items**
- **append(x):** Adds one item at the **end**.
- **insert(i, x):** Adds an item at **position** `i`.
- **extend(list):** Adds **many** items at the end.

```python
trip = ["Siliguri"]
trip.append("Darjeeling")
trip.insert(1, "Kurseong")
trip.extend(["Kalimpong", "Lava", "Lolegaon"])
print(trip)
```

```python
# Difference between append and extend
a = [1, 2]
a.append([3, 4])      # Adds the whole list as ONE item
print(a)

b = [1, 2]
b.extend([3, 4])      # Adds each item separately
print(b)
```

## **Removing Items**
- **remove(x):** Removes the first item with value `x`.
- **pop(i):** Removes the item at position `i` and gives it back. Without `i`, it removes the last item.
- **del list[i]:** Deletes the item at position `i`.
- **clear():** Removes everything.

```python
trip.remove("Lava")
print(trip)

last = trip.pop()
print("Removed:", last, "->", trip)

first = trip.pop(0)
print("Removed:", first, "->", trip)

del trip[0]
print(trip)

# trip.remove("Goa")      # This will raise a ValueError (Goa is not in the list)
```

## **Searching and Counting**

```python
fish_orders = ["ilish", "rui", "ilish", "katla", "ilish", "pabda", "rui"]

print(fish_orders.count("ilish"))     # How many times?
print(fish_orders.index("katla"))     # At which position?
print("bhetki" in fish_orders)        # Is it there?
```

## **Sorting and Reversing**
- **sort():** Sorts the list itself (changes it).
- **sorted():** Gives a **new** sorted list and keeps the original unchanged.
- **reverse():** Reverses the order.

```python
cities = ["Siliguri", "Asansol", "Kolkata", "Durgapur", "Bardhaman", "Howrah"]

cities.sort()
print(cities)

cities.sort(reverse=True)
print(cities)

cities.sort(key=len)            # Sort by the length of the name
print(cities)

cities.reverse()
print(cities)
```

```python
# Common mistake: sort() returns None
x = [3, 1, 2]
y = x.sort()
print(y)        # None, because sort() changes x itself and returns nothing
print(x)
```

## **Sorting a List of Lists by One Column**

```python
# District, population (Census 2011)
data = [
    ["Kolkata", 4496694],
    ["North 24 Parganas", 10009781],
    ["Paschim Medinipur", 5913457],
    ["Darjeeling", 1846823],
    ["Howrah", 4850029],
]

data.sort(key=lambda row: row[1], reverse=True)
for name, pop in data:
    print(f"{name:<20}{pop:>12,}")
```

- `key=lambda row: row[1]`: This tells `sort()` to use the **second value** (population) of each row for sorting. `lambda` is a small one-line function, which we will learn on Day 24.

## **Example from Remote Sensing: Choosing the Best Images**
We searched for satellite images of Kolkata and got a list of dates with their cloud cover. Let us sort them from the clearest to the cloudiest, and pick the best three.

```python
scenes = [
    ["2024-01-05", 12],
    ["2024-01-15", 3],
    ["2024-01-25", 45],
    ["2024-02-04", 0],
    ["2024-02-14", 22],
    ["2024-02-24", 7],
]

scenes.sort(key=lambda s: s[1])
print("All scenes (clearest first):")
for date, cloud in scenes:
    print(f"  {date}  cloud {cloud:>2}%")

best = scenes[:3]
print("Best 3 dates:", [s[0] for s in best])
```

## **Using a List as a Stack and a Queue**
- **Stack:** Last In, First Out (like a pile of plates at a wedding).
- **Queue:** First In, First Out (like a line at a ration shop).

```python
# Stack
stack = []
stack.append("plate1")
stack.append("plate2")
stack.append("plate3")
print(stack.pop(), stack.pop())      # The last one comes out first

# Queue
queue = ["Ram", "Shyam", "Jodu", "Modhu"]
print("Served:", queue.pop(0))       # The first one comes out first
queue.append("Bishu")
print(queue)
```

## **Practice Problems**
1. Remove duplicate items from a list, but keep the original order.
2. Find the second largest number without sorting.

```python
visited = ["Digha", "Mandarmani", "Digha", "Tajpur", "Mandarmani", "Shankarpur"]
unique = []
for p in visited:
    if p not in unique:
        unique.append(p)
print(unique)
```

```python
marks = [67, 88, 92, 75, 92, 81]
first = second = float("-inf")
for m in marks:
    if m > first:
        first, second = m, first
    elif first > m > second:
        second = m
print("Largest:", first, " Second largest:", second)
```

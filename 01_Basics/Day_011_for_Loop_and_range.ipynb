# **Day 11: The for Loop and range()**

The `for` loop goes through a group of items **one by one**. For each item, it runs the code inside the loop.

**When to use which loop?**
- **for loop:** When we know what we are looping over (a list, a string, a range of numbers).
- **while loop:** When we only know the stopping condition.

```python
for city in ["Kolkata", "Howrah", "Durgapur", "Asansol", "Siliguri"]:
    print(city)
```

```python
# Looping over a string gives one character at a time
for ch in "ILISH":
    print(ch, end=" ")
```

## **The range() Function**
`range()` makes a sequence of numbers. It is mostly used with `for` loops.

| Form | Meaning | Example | Numbers |
|---|---|---|---|
| `range(stop)` | 0 up to stop-1 | `range(5)` | 0, 1, 2, 3, 4 |
| `range(start, stop)` | start up to stop-1 | `range(1, 6)` | 1, 2, 3, 4, 5 |
| `range(start, stop, step)` | jump by step | `range(0, 21, 5)` | 0, 5, 10, 15, 20 |

**Note:** The `stop` number is never included.

```python
print(list(range(5)))
print(list(range(1, 6)))
print(list(range(0, 21, 5)))
print(list(range(10, 0, -2)))     # Counting backwards
```

```python
for i in range(1, 6):
    print(i, i ** 2, i ** 3)
```

## **enumerate(): Getting the Position Also**
`enumerate()` gives us both the **position number** and the **item** in each round.

```python
stations = ["Esplanade", "Chandni Chowk", "Central", "Girish Park", "Shobhabazar"]

for number, station in enumerate(stations, start=1):
    print(number, station)
```

## **Example: Monthly Rainfall of Kolkata**
(Approximate monthly rainfall in mm, only for practice.)

```python
rain = [11, 30, 35, 60, 140, 300, 390, 350, 330, 160, 25, 5]
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

total = 0
for r in rain:
    total += r

print("Total rainfall:", total, "mm")
print("Average per month:", round(total / len(rain), 1), "mm")
```

```python
# Find the wettest month without using max()
wettest = 0
for i in range(len(rain)):
    if rain[i] > rain[wettest]:
        wettest = i

print("Wettest month:", months[wettest], "with", rain[wettest], "mm")
```

## **zip(): Looping Over Two Lists Together**
`zip()` joins two lists side by side, so we get one item from each list in every round.

```python
for m, r in zip(months, rain):
    bar = "#" * (r // 20)
    print(f"{m} {r:>4} {bar}")
```

This makes a small bar chart using `#`. We can clearly see the **monsoon** from June to September.

## **Example from Remote Sensing: Looping Over Satellite Bands**
A satellite does not take one photo. It takes many "photos" of the same place, each in a different colour of light. Each such photo is called a **band**.

Some important **Landsat 8** bands:

| Band | Name | Used for |
|---|---|---|
| 2 | Blue | Water, coastal areas |
| 3 | Green | Vegetation, water |
| 4 | Red | Vegetation, soil |
| 5 | Near Infrared (NIR) | Healthy vegetation (plants reflect a lot of NIR) |
| 6 | Shortwave Infrared (SWIR) | Soil moisture, burnt areas |

```python
band_names = ["Blue", "Green", "Red", "NIR", "SWIR"]
band_numbers = [2, 3, 4, 5, 6]

for number, name in zip(band_numbers, band_names):
    print(f"Band {number}: {name}")
```

```python
# Reflectance of one forest pixel in each band (values from 0 to 1)
reflectance = [0.04, 0.07, 0.05, 0.45, 0.20]

for name, value in zip(band_names, reflectance):
    bar = "#" * int(value * 50)
    print(f"{name:<6}{value:>5}  {bar}")
```

Healthy plants reflect very little red light but a lot of **NIR** light. You can see this clearly in the output: the NIR bar is the longest. This idea is the base of NDVI.

## **Example: Factorial and Fibonacci**

```python
# Factorial: 6! = 6 x 5 x 4 x 3 x 2 x 1
n = 6
fact = 1
for i in range(1, n + 1):
    fact *= i
print(f"{n}! = {fact}")
```

```python
# Fibonacci: each number is the sum of the two before it
a, b = 0, 1
for _ in range(12):
    print(a, end=" ")
    a, b = b, a + b
```

- `_` is used as the loop variable when we do not need its value.

## **Practice Problems**
1. Count the vowels in "Murshidabad".
2. Check if 97 is a prime number.

```python
word = "Murshidabad"
count = 0
for ch in word.lower():
    if ch in "aeiou":
        count += 1
print("Vowels:", count)
```

```python
n = 97
is_prime = n > 1
for i in range(2, int(n ** 0.5) + 1):
    if n % i == 0:
        is_prime = False
        break
print(n, "is prime:", is_prime)
```

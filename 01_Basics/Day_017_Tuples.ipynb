# **Day 17: Tuples**

A **tuple** is like a list, but it **cannot be changed** after it is made. It uses round brackets `( )`.

**When should we use a tuple?** When the data should stay fixed. A very good example is a **coordinate** (latitude, longitude): the location of Kolkata does not change.

## **Creating a Tuple**

```python
kolkata = (22.5726, 88.3639)
print(kolkata)
print(type(kolkata))
print(kolkata[0])

# kolkata[0] = 23.0      # This will raise a TypeError (tuples cannot be changed)
```

```python
# A tuple with one item needs a comma
a = ("Howrah")         # This is just a string
b = ("Howrah",)        # This is a tuple
print(type(a), type(b))

# Brackets are optional
c = 1, 2, 3
print(type(c))
```

## **Unpacking a Tuple**
**Unpacking** means taking the values out of a tuple into separate variables in one line.

```python
lat, lon = kolkata
print("Latitude:", lat, " Longitude:", lon)

place, pin, state = ("Kharagpur", 721301, "WB")
print(place, pin, state)
```

```python
# Use * to collect the remaining values
first, *middle, last = ("Howrah", "Bally", "Uttarpara", "Konnagar", "Rishra", "Serampore")
print(first)
print(middle)
print(last)
```

## **Tuple Methods**
Tuples have only two methods, because they cannot be changed.

```python
t = (1, 2, 2, 3, 2)
print(t.count(2))      # How many times 2 appears
print(t.index(3))      # Position of 3
```

## **Why Use a Tuple Instead of a List?**
1. **Safety:** Nobody can change the data by mistake.
2. **Speed and memory:** Tuples are a little faster and smaller.
3. **Dictionary keys:** Tuples can be used as keys in a dictionary, lists cannot.

```python
import sys
print(sys.getsizeof([1, 2, 3, 4, 5]), "bytes for the list")
print(sys.getsizeof((1, 2, 3, 4, 5)), "bytes for the tuple")
```

```python
# Tuple as a dictionary key: distance between two cities
distance = {
    ("Kolkata", "Siliguri"): 560,
    ("Kolkata", "Digha"): 185,
    ("Kolkata", "Midnapore"): 130,
}
print(distance[("Kolkata", "Digha")], "km (approx road distance)")
```

## **Example from GIS: Points of Interest**
Let us store some places with their coordinates as a **list of tuples**. Each tuple is `(name, latitude, longitude)`.

```python
places = [
    ("Victoria Memorial", 22.5448, 88.3426),
    ("Howrah Bridge", 22.5851, 88.3468),
    ("Tiger Hill", 27.0126, 88.2630),
    ("Digha Beach", 21.6266, 87.5074),
    ("Bishnupur Temples", 23.0747, 87.3197),
]

for name, lat, lon in places:
    region = "North Bengal" if lat > 24.5 else "South Bengal"
    print(f"{name:<18} ({lat:.4f}, {lon:.4f})  {region}")
```

- `for name, lat, lon in places`: Each tuple is unpacked into three variables in every round of the loop.
- Latitude above about 24.5 is roughly North Bengal, which is a simple rule for this example.

```python
# Find the northernmost place
northmost = max(places, key=lambda p: p[1])
print("Northernmost place:", northmost[0])
```

## **A Function Returning Many Values Gives a Tuple**

```python
def min_max(nums):
    return min(nums), max(nums)

result = min_max([34, 12, 56, 7])
print(result, type(result))

low, high = min_max([34, 12, 56, 7])
print(low, high)
```

## **Practice Problem**
Sort these famous places by the year they were built.

```python
monuments = [("Victoria Memorial", 1921), ("Howrah Bridge", 1943), ("Writers' Building", 1777)]

for name, year in sorted(monuments, key=lambda m: m[1]):
    print(f"{name} - {year}")
```

# **Day 3: Numbers and Operators**

Python can work like a calculator. The symbols we use for calculation are called **operators**.

## **Arithmetic Operators**

| Operator | Name | Example | Result |
|---|---|---|---|
| `+` | Addition | `10 + 3` | `13` |
| `-` | Subtraction | `10 - 3` | `7` |
| `*` | Multiplication | `10 * 3` | `30` |
| `/` | Division | `10 / 3` | `3.333...` |
| `//` | Floor division (no decimal part) | `10 // 3` | `3` |
| `%` | Modulus (remainder) | `10 % 3` | `1` |
| `**` | Power | `10 ** 3` | `1000` |

```python
print(10 + 3)
print(10 - 3)
print(10 * 3)
print(10 / 3)     # Division always gives a float
print(10 // 3)    # Floor division
print(10 % 3)     # Remainder
print(10 ** 3)    # Power
```

## **Example: Splitting a Phuchka Bill**
Four friends ate phuchka at Vivekananda Park. The total bill is Rs 170.

```python
bill = 170
friends = 4

print("Each pays (exact):", bill / friends)
print("Each pays (whole rupees):", bill // friends)
print("Rupees left over:", bill % friends)
```

## **Example from Remote Sensing: Pixel Area**
In a satellite image, every pixel covers a square area on the ground.
- A **Landsat** pixel is 30 m x 30 m.
- A **Sentinel-2** pixel is 10 m x 10 m.

Let us calculate the ground area of a pixel, and the area of a water body made of many pixels.

```python
landsat_pixel = 30
sentinel_pixel = 10

landsat_area = landsat_pixel ** 2       # Area of one pixel in square metre
sentinel_area = sentinel_pixel ** 2

print("One Landsat pixel covers", landsat_area, "square metre")
print("One Sentinel-2 pixel covers", sentinel_area, "square metre")

# A pond in a Landsat image is made of 250 pixels
pond_pixels = 250
pond_area_m2 = pond_pixels * landsat_area
pond_area_hectare = pond_area_m2 / 10000     # 1 hectare = 10,000 square metre

print("Pond area:", pond_area_m2, "square metre =", pond_area_hectare, "hectare")
```

- `landsat_pixel ** 2`: This means 30 x 30 = 900 square metre.
- `pond_area_m2 / 10000`: We divide by 10,000 to change square metre into hectare.

## **Order of Operations (BODMAS)**
Python follows the same rules as school mathematics:
1. Brackets `()`
2. Power `**`
3. Multiplication and division `* / // %`
4. Addition and subtraction `+ -`

If you are not sure, use brackets. They make the code easier to read too.

```python
print(2 + 3 * 4)       # Multiplication first
print((2 + 3) * 4)     # Brackets first
print(2 ** 3 ** 2)     # Power goes from right to left: 3**2 = 9, then 2**9
```

## **Assignment Operators**
These are short forms for updating a variable.

| Short form | Same as |
|---|---|
| `x += 5` | `x = x + 5` |
| `x -= 5` | `x = x - 5` |
| `x *= 5` | `x = x * 5` |
| `x /= 5` | `x = x / 5` |

```python
metro_card_balance = 200
metro_card_balance -= 20     # Dum Dum to Esplanade
metro_card_balance -= 20     # Return journey
metro_card_balance += 100    # Recharge
print("Balance:", metro_card_balance)
```

## **Converting and Rounding Numbers**

```python
print(int(9.99))          # int() removes the decimal part
print(float(7))           # float() adds a decimal part
print(round(9.99))        # round() rounds to the nearest whole number
print(round(3.14159, 2))  # Round to 2 decimal places
print(abs(-40))           # abs() gives the positive value
```

## **A Small Surprise with Decimals**

```python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
```

Why is it not exactly 0.3? The computer stores decimal numbers in **binary** (0s and 1s), and some decimals like 0.1 cannot be stored exactly. It is not a Python mistake. For money, we can use `round()`.

```python
print(round(0.1 + 0.2, 2))
```

## **The math Module**
For more maths functions, Python gives us the `math` module.

```python
import math

print(math.sqrt(144))     # Square root
print(math.pi)            # Value of pi
print(math.ceil(4.1))     # Round up
print(math.floor(4.9))    # Round down
print(math.factorial(5))  # 5 x 4 x 3 x 2 x 1
```

## **Practice Problems**
1. A round pond near Bankura has a radius of 12.5 m. Find its area and boundary length.
2. A map has a scale of **1:50,000**. This means 1 cm on the map = 50,000 cm on the ground. If two villages are 4.2 cm apart on the map, what is the real distance in km?

```python
# Solution 1
r = 12.5
area = math.pi * r ** 2
boundary = 2 * math.pi * r
print("Area:", round(area, 2), "square metre")
print("Boundary:", round(boundary, 2), "metre")
```

```python
# Solution 2
map_distance_cm = 4.2
scale = 50000

ground_cm = map_distance_cm * scale
ground_km = ground_cm / 100 / 1000      # cm -> m -> km
print("Real distance:", ground_km, "km")
```

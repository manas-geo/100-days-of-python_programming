# **Day 8: Conditions with if, elif and else**

In real life we make decisions all the time: **if** it rains, take an umbrella, **else** take sunglasses. In Python, we make decisions using `if`, `elif` and `else`.

**Important rule:** The code inside `if` must be moved to the right (called **indentation**). The standard is 4 spaces. Python uses this space to understand which lines belong to the `if`.

## **The if Statement**

```python
temp = 38

if temp > 35:
    print("Too hot, stay inside")

print("This line always runs")
```

## **if-else**

```python
temp = 28

if temp > 35:
    print("Too hot")
else:
    print("Nice weather to go out")
```

## **if-elif-else**
`elif` means "else if". We use it when there are more than two choices. Python checks the conditions **from top to bottom** and runs only the **first** one that is True.

```python
temp = 12

if temp >= 35:
    print("Kolkata summer")
elif temp >= 25:
    print("Normal day")
elif temp >= 15:
    print("Pleasant, like Puja time")
else:
    print("Cold like Darjeeling, take a jacket")
```

## **Example: Grade Calculator**

```python
marks = 82

if marks >= 90:
    grade = "AA"
elif marks >= 80:
    grade = "A+"
elif marks >= 60:
    grade = "A"
elif marks >= 45:
    grade = "B"
elif marks >= 35:
    grade = "C"
else:
    grade = "Fail"

print("Grade:", grade)
```

**Note:** The order of conditions matters. If we check `marks >= 35` first, then 95 marks will also get grade "C", because the first True condition wins.

## **Example from Remote Sensing: Classifying NDVI**
**NDVI** (Normalized Difference Vegetation Index) is a number that tells us how green (healthy) the vegetation is at a place. It is calculated from a satellite image and always lies between **-1 and +1**.

A simple way to read NDVI:

| NDVI value | What it usually means |
|---|---|
| below 0 | Water |
| 0 to 0.2 | Bare soil, sand, buildings |
| 0.2 to 0.5 | Sparse vegetation (grass, crops) |
| above 0.5 | Dense vegetation (forest) |

We will learn how to calculate NDVI later. Today, let us classify an NDVI value using `if-elif-else`.

```python
ndvi = 0.72       # For example, a pixel in the Sundarbans forest

if ndvi < 0:
    land_cover = "Water"
elif ndvi < 0.2:
    land_cover = "Bare soil / Built-up"
elif ndvi < 0.5:
    land_cover = "Sparse vegetation"
else:
    land_cover = "Dense vegetation"

print("NDVI:", ndvi, "->", land_cover)
```

## **Example: Electricity Bill (Slab System)**
Electricity bills are calculated in slabs. (These rates are only for practice.)
- First 100 units: Rs 5 per unit
- Next 100 units: Rs 6 per unit
- Above 200 units: Rs 7.5 per unit

```python
units = 260

if units <= 100:
    bill = units * 5
elif units <= 200:
    bill = 100 * 5 + (units - 100) * 6
else:
    bill = 100 * 5 + 100 * 6 + (units - 200) * 7.5

print(f"Units: {units}, Bill: Rs {bill}")
```

## **One Line if-else (Ternary Operator)**
For simple cases, we can write `if-else` in one line:

`value_if_true if condition else value_if_false`

```python
age = 17
status = "Can vote" if age >= 18 else "Cannot vote yet"
print(status)
```

## **Practice Problems**
1. Find which part of West Bengal a district belongs to.
2. Check if a number is odd or even.

```python
district = "Siliguri"

north = ["Darjeeling", "Siliguri", "Jalpaiguri", "Cooch Behar", "Alipurduar", "Kalimpong"]
south = ["Kolkata", "Howrah", "South 24 Parganas", "Purba Medinipur", "Paschim Medinipur"]

if district in north:
    print(district, "is in North Bengal")
elif district in south:
    print(district, "is in South Bengal")
else:
    print("Not in the list")
```

```python
n = 721
if n % 2 == 0:
    print(n, "is even")
else:
    print(n, "is odd")
```

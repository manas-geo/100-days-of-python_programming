# **Day 12: break, continue, pass and else with Loops**

These keywords give us more control over loops:

| Keyword | What it does |
|---|---|
| `break` | Stops the loop **completely** |
| `continue` | Skips the **rest of this round** and goes to the next round |
| `pass` | Does **nothing**. It is used as an empty placeholder |

## **break**

```python
route = ["Sealdah", "Bidhannagar", "Dum Dum", "Belgharia", "Agarpara", "Sodepur", "Barrackpore"]

for station in route:
    print("At", station)
    if station == "Sodepur":
        print("Getting down here")
        break
```

## **continue**

```python
# Skip the stations where this local train does not stop
no_stop = {"Belgharia", "Agarpara"}

for station in route:
    if station in no_stop:
        continue
    print("Stops at", station)
```

## **pass**
Python does not allow an empty block. `pass` is used when we want to write the code later.

```python
for station in route:
    if station == "Dum Dum":
        pass    # To do: add metro change information here
print("Loop finished without any error")
```

## **else with a Loop**
A loop can have an `else` part. The `else` runs **only if the loop was not stopped by `break`**.

This is very useful for searching: if we did not find the item (no `break`), the `else` part tells us so.

```python
districts = ["Kolkata", "Howrah", "Hooghly"]
target = "Nadia"

for d in districts:
    if d == target:
        print("Found", target)
        break
else:
    print(target, "not found")
```

```python
# Prime numbers from 2 to 50 using for-else
for n in range(2, 51):
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            break
    else:
        print(n, end=" ")
```

Let us understand the code above:
- For each number `n`, the inner loop checks if any number divides it.
- If a divisor is found, `break` stops the inner loop, and the `else` does **not** run.
- If no divisor is found, the inner loop finishes normally and the `else` prints `n` as a prime.

## **Example from Remote Sensing: Skipping Bad Pixels**
Sometimes a pixel has no data (for example, because of a cloud or an edge of the image). Such pixels are often stored with a special value like **-9999**, called the **NoData value**. When calculating an average, we must **skip** them.

```python
pixel_values = [0.32, 0.41, -9999, 0.38, -9999, 0.45, 0.29]
NODATA = -9999

total = 0
count = 0
for v in pixel_values:
    if v == NODATA:
        continue          # Skip this pixel
    total += v
    count += 1

print("Valid pixels:", count)
print("Average value:", round(total / count, 3))
```

If we did not skip -9999, the average would become a very large negative number, which is completely wrong.

## **Example: ATM PIN with 3 Attempts**

```python
correct_pin = "4321"
attempts = ["1111", "1234", "4321"]    # Suppose the user typed these

for tries, pin in enumerate(attempts, start=1):
    if pin == correct_pin:
        print(f"Welcome! (attempt {tries})")
        break
    print("Wrong PIN")
else:
    print("Card blocked")
```

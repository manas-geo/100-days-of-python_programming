# **Day 4: Python Strings**

A **string** is text. It is a sequence (line) of characters, like letters, numbers, spaces and symbols.

We can write strings inside:
- single quotes `'...'`
- double quotes `"..."`
- triple quotes `'''...'''` or `"""..."""` for text with many lines

## **Creating Strings**

````python
# Single line strings
s1 = 'Kolkata'
s2 = "Siliguri"

# Multi line string
s3 = """Darjeeling
is the queen
of hills"""

print(s1)
print(s2)
print(s3)
````

````python
# Double quotes are useful when the text has an apostrophe
print("It's raining in Kolkata.")
````

## **Accessing Characters (Indexing)**
Each character in a string has a position number called an **index**.
- **Positive indexing** starts from `0` at the left.
- **Negative indexing** starts from `-1` at the right (last character).

```
 K   o   l   k   a   t   a
 0   1   2   3   4   5   6
-7  -6  -5  -4  -3  -2  -1
```

````python
city = "Kolkata"

# Positive indexing
print(city[0])
print(city[3])

# Negative indexing
print(city[-1])
print(city[-2])

print(len(city))    # len() gives the number of characters

# print(city[10])   # This will raise an IndexError (there is no position 10)
````

## **Slicing**
Slicing means taking out a part of a string. The format is `string[start:stop:step]`.
- `start`: where to begin (included)
- `stop`: where to end (**not** included)
- `step`: how many positions to jump

````python
place = "Midnapore"

print(place[0:3])     # From index 0 to 2
print(place[3:])      # From index 3 to the end
print(place[:4])      # From start to index 3
print(place[-4:])     # Last 4 characters
print(place[::2])     # Every second character
print(place[::-1])    # Reverse the string
````

## **Example from Remote Sensing: Reading a Landsat Scene ID**
Every Landsat image has a long name called the **Scene ID**. Each part of the name has a meaning.

`LC08_L1TP_138044_20231110_20231115_02_T1`

| Part | Meaning |
|---|---|
| `LC08` | Landsat 8 |
| `L1TP` | Processing level |
| `138044` | Path 138, Row 044 (the location of the image on Earth) |
| `20231110` | Date the image was taken (10 November 2023) |

Using slicing, we can pull out each part.

````python
scene_id = "LC08_L1TP_138044_20231110_20231115_02_T1"

satellite = scene_id[0:4]
path = scene_id[10:13]
row = scene_id[13:16]
year = scene_id[17:21]
month = scene_id[21:23]
day = scene_id[23:25]

print("Satellite:", satellite)
print("Path:", path, " Row:", row)
print("Date taken:", day + "/" + month + "/" + year)
````

Let us understand the code above:
- `scene_id[0:4]`: Takes characters from index 0 to 3, which is `LC08`.
- `scene_id[17:21]`: Takes the year part `2023`.
- `day + "/" + month + "/" + year`: Joins the pieces together to make a date.

## **Strings Cannot Be Changed (Immutable)**
Once a string is created, we **cannot change** a single character inside it. But we can make a new string.

````python
name = "Howra"
# name[4] = "h"        # This will raise a TypeError (strings are immutable)

name = name + "h"      # Making a new string is allowed
print(name)

# We can delete the whole string
del name
# print(name)          # This will raise a NameError (name no longer exists)
````

## **Operations on Strings**

**1. Concatenation (+):** Joins two strings.

**2. Repetition (*):** Repeats a string many times.

````python
first = "Durga"
second = "Puja"
print(first + " " + second)
print("Dhak " * 3)

# print("PIN " + 700001)        # This will raise a TypeError (cannot add str and int)
print("PIN " + str(700001))     # Convert the number to a string first
````

**3. Membership (in / not in):** Checks if some text is inside a string.

````python
sweets = "rosogolla sandesh mishti doi langcha"

print("sandesh" in sweets)
print("pantua" in sweets)
print("pantua" not in sweets)
````

**4. Loop over a string:** We can go through each character one by one (loops are covered on Day 11).

````python
for ch in "Digha":
    print(ch)
````

**5. Comparison:** Strings can be compared. Capital letters and small letters are different.

````python
print("Kolkata" == "Kolkata")
print("Kolkata" == "kolkata")
````

## **Palindrome Check**
A **palindrome** is a word that reads the same from both sides, like "malayalam".

````python
word = "malayalam"
print(word == word[::-1])

word = "kolkata"
print(word == word[::-1])
````

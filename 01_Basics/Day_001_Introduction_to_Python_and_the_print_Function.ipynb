## **The print() Function**
The `print()` function shows something on the screen. Whatever we put inside the brackets gets printed.

Text must be written inside quotes (`" "` or `' '`). Numbers can be written without quotes.

```python
# Printing text
print("Hello World")
print("Hello Kolkata")
```

```python
# Printing numbers and True/False values
print(2024)
print(5.9)
print(True)
```

## **Printing Many Values Together**
We can print many values in one `print()` by separating them with commas. Python automatically puts a **space** between them.

```python
print("Kolkata", "Howrah", "Siliguri")
print("PIN code of Midnapore town is", 721101)
print("Landsat 8 image resolution is", 30, "metre")
```

## **The sep and end Parameters**
- **sep:** What to put **between** the values. By default it is a space.
- **end:** What to put at the **end** of the line. By default it is a new line (`\n`), so the next print starts on a new line.

```python
# Using sep
print("Kolkata", "Howrah", "Hooghly", sep=" -> ")
print("21", "02", "1952", sep="/")
```

```python
# Without end, each print goes to a new line
print("Rosogolla")
print("Mishti Doi")

# With end, prints stay on the same line
print("Rosogolla", end=" | ")
print("Mishti Doi", end=" | ")
print("Sandesh")
```

Let us understand the code above:
- `print("Rosogolla", end=" | ")`: After printing "Rosogolla", Python prints ` | ` instead of going to the next line.
- So all three sweets are printed on one single line.

## **Comments**
A **comment** is a note for humans. Python ignores it completely.
- A single line comment starts with `#`.
- For longer notes, we can use triple quotes `""" """`.

Comments help us (and others) understand the code later.

```python
# This is a single line comment
print("Comments are not printed")  # A comment can also be written after code

"""
This is a multi line note.
Python simply ignores it.
"""
print("Done")
```

## **Escape Characters**
Some characters cannot be typed directly inside a string. For them we use a backslash `\`.

| Escape Character | Meaning |
|---|---|
| `\n` | New line |
| `\t` | Tab (big space) |
| `\"` | Double quote inside a string |
| `\\` | A real backslash |

```python
# New line
print("Howrah\nSealdah\nShalimar")
```

```python
# Tab space makes a neat table
print("Station\tCode")
print("Howrah\tHWH")
print("Sealdah\tSDAH")
print("Kharagpur\tKGP")
```

```python
# Quotes and backslash inside a string
print("He said \"Cha khabe?\"")
print('Or use single quotes outside: "Cha khabe?"')
print("File path: C:\\GIS\\data\\kolkata.tif")
```

## **A Small Example: Satellite Information Board**
Let us print a small information board about a satellite using only what we learned today.

```python
print("=" * 40)
print("        SATELLITE INFORMATION")
print("=" * 40)
print("Name", "Sentinel-2", sep="\t\t: ")
print("Agency", "European Space Agency", sep="\t\t: ")
print("Resolution", "10 metre", sep="\t: ")
print("Revisit time", "5 days", sep="\t: ")
print("=" * 40)
```

- `"=" * 40`: This repeats the `=` sign 40 times to make a line. We will learn more about this on Day 4.

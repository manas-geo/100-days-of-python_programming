## **Type Conversion**
**Type conversion** means changing a value from one data type to another.

There are two kinds:
1. **Explicit conversion:** We do it ourselves using functions like `int()`, `float()`, `str()`, `bool()`.
2. **Implicit conversion:** Python does it automatically when it is safe.

### **Explicit Conversion**

```python
age = int("25")
print(age + age)

price = float("45.50")
print(price * 2)

pin = str(721301)
print("Kharagpur PIN: " + pin)
```

```python
# Some conversions are not possible
# int("kolkata")    # This will raise a ValueError (text is not a number)
# int("45.5")       # This will raise a ValueError (int() cannot read a decimal string)

print(int(float("45.5")))    # First convert to float, then to int
```

### **Implicit Conversion**

```python
x = 10 + 2.5          # int + float gives float
print(x, type(x))

print(True + True)    # True is treated as 1, False as 0
```

## **Converting to Boolean**
With `bool()`:
- **Empty** things and **zero** become `False`.
- Everything else becomes `True`.

```python
print(bool(0), bool(1), bool(-5))
print(bool(""), bool("hi"))
print(bool([]), bool([0]))
print(bool(None))
```

## **Taking Many Values in One Line**
Sometimes the user types many values in one line, separated by spaces. We can use `split()` and then convert each part.

```python
# a, b = input("Enter two numbers: ").split()

line = "12 30"                      # Suppose the user typed this
a, b = line.split()
a, b = int(a), int(b)
print(a + b)

# Shorter way using map()
a, b = map(int, "12 30".split())
print(a + b)
```

## **Example from GIS: Entering a Location**
A GIS user types the latitude and longitude of a place. Let us convert them into numbers and print them nicely.

- **Latitude** tells how far north or south a place is from the equator.
- **Longitude** tells how far east or west a place is from Greenwich (London).

```python
# lat_text = input("Latitude: ")
# lon_text = input("Longitude: ")

lat_text = "22.4257"       # Midnapore (suppose the user typed this)
lon_text = "87.3199"

lat = float(lat_text)
lon = float(lon_text)

print(f"Location: {lat:.4f} N, {lon:.4f} E")
print(type(lat), type(lon))
```

## **Example: Yellow Taxi Fare**
(The fare values here are only for practice.)

```python
km = float("7.5")          # For example, Sealdah to Salt Lake
base_fare = 30
per_km = 25

fare = base_fare + max(0, km - 2) * per_km
print(f"Approximate fare: Rs {fare:.0f}")
```

- `max(0, km - 2)`: The first 2 km are included in the base fare, so we charge only for the rest. `max()` makes sure the value never goes below 0.

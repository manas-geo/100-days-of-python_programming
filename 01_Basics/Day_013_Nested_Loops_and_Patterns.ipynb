# **Day 13: Nested Loops and Patterns**

A **nested loop** is a loop inside another loop. For **every single round** of the outer loop, the inner loop runs **completely**.

Think of a clock: for every 1 round of the hour hand, the minute hand goes round 60 times.

```python
for i in range(1, 4):
    for j in range(1, 4):
        print(f"({i},{j})", end=" ")
    print()      # Go to a new line after the inner loop
```

## **Pattern Printing**
Patterns are a good way to understand how nested loops work.

```python
# Right triangle
n = 5
for i in range(1, n + 1):
    for j in range(i):
        print("*", end=" ")
    print()
```

```python
# Pyramid
n = 5
for i in range(1, n + 1):
    print(" " * (n - i) + "* " * i)
```

```python
# Number triangle
for i in range(1, 6):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()
```

```python
# Diamond
n = 4
for i in range(1, n + 1):
    print(" " * (n - i) + "*" * (2 * i - 1))
for i in range(n - 1, 0, -1):
    print(" " * (n - i) + "*" * (2 * i - 1))
```

## **Example from Remote Sensing: An Image is a Grid of Pixels**
A satellite image is a **grid** (table) of pixels. It has **rows** (going down) and **columns** (going across). To visit every pixel, we need two loops:
- the **outer loop** goes through each row
- the **inner loop** goes through each column in that row

Below is a tiny 4 x 5 image. Each number is the **NDVI** of one pixel.

```python
image = [
    [0.10, 0.15, 0.62, 0.70, 0.68],
    [0.12, 0.45, 0.66, 0.72, 0.71],
    [-0.20, -0.25, 0.30, 0.55, 0.60],
    [-0.30, -0.28, 0.05, 0.08, 0.40],
]

rows = len(image)
cols = len(image[0])
print("Image size:", rows, "rows x", cols, "columns")

# Show each pixel as a symbol: ~ water, . bare, + sparse, # dense vegetation
for r in range(rows):
    for c in range(cols):
        v = image[r][c]
        if v < 0:
            symbol = "~"
        elif v < 0.2:
            symbol = "."
        elif v < 0.5:
            symbol = "+"
        else:
            symbol = "#"
        print(symbol, end=" ")
    print()
```

Let us understand the code above:
- `image[r][c]`: Gets the pixel in row `r` and column `c`.
- The inner loop prints one full row. Then `print()` moves to the next line.
- The result looks like a small map: water (`~`) at the bottom left, and dense vegetation (`#`) on the right.

```python
# Count how many pixels are dense vegetation
dense = 0
for r in range(rows):
    for c in range(cols):
        if image[r][c] >= 0.5:
            dense += 1

total = rows * cols
print(f"Dense vegetation: {dense} of {total} pixels ({dense / total:.0%})")
```

## **Example: Football Fixtures**
Every team plays every other team once.

```python
teams = ["Mohun Bagan", "East Bengal", "Mohammedan", "Bhawanipore"]
match_no = 1

for i in range(len(teams)):
    for j in range(i + 1, len(teams)):
        print(f"Match {match_no}: {teams[i]} vs {teams[j]}")
        match_no += 1
```

## **Practice Problem: Hollow Square**

```python
n = 6
for i in range(n):
    for j in range(n):
        if i == 0 or i == n - 1 or j == 0 or j == n - 1:
            print("#", end=" ")
        else:
            print(" ", end=" ")
    print()
```

# **Day 10: The while Loop**

A **loop** repeats some code again and again. The `while` loop keeps repeating **as long as a condition is True**.

**Be careful:** Something inside the loop must change, so that the condition becomes False one day. Otherwise the loop never stops. This is called an **infinite loop**.

```python
i = 1
while i <= 5:
    print("Dhak", i)
    i += 1          # Increase i, otherwise the loop never ends
```

Let us understand the code above:
- `i = 1`: We start counting from 1.
- `while i <= 5`: The loop runs as long as `i` is 5 or less.
- `i += 1`: After each round, `i` increases by 1. When `i` becomes 6, the loop stops.

## **Example: Sum of Numbers from 1 to n**

```python
n = 100
total = 0
i = 1

while i <= n:
    total += i
    i += 1

print("Sum using loop:", total)
print("Sum using formula:", n * (n + 1) // 2)
```

## **Example: Countdown for Bisarjan**

```python
count = 5
while count > 0:
    print(count)
    count -= 1
print("Asche bochor abar hobe!")
```

## **Example: Reverse a Number and Sum its Digits**
- `num % 10` gives the **last digit**.
- `num // 10` **removes** the last digit.

```python
num = 721302
reverse = 0
digit_sum = 0
temp = num

while temp > 0:
    digit = temp % 10
    reverse = reverse * 10 + digit
    digit_sum += digit
    temp //= 10

print("Reverse:", reverse)
print("Sum of digits:", digit_sum)
```

## **Example: Saving for a Darjeeling Trip**
We need Rs 15,000 for a trip. We save Rs 1,200 every month, and the bank adds 1% every month (an imaginary bank). How many months will it take?

```python
target = 15000
saved = 0
months = 0

while saved < target:
    saved += 1200
    saved *= 1.01
    months += 1

print(f"{months} months, saved Rs {saved:.2f}")
```

## **Example from Remote Sensing: Finding a Cloud-Free Image**
Suppose we have a list of cloud cover values for images taken on different dates. We want the **first** image with cloud cover below 10%. We keep checking images **while** we have not found a good one.

```python
dates = ["05-Jun", "10-Jun", "15-Jun", "20-Jun", "25-Jun", "30-Jun"]
cloud = [85, 72, 64, 40, 9, 15]      # Monsoon, so very cloudy at first

i = 0
while i < len(cloud) and cloud[i] >= 10:
    print(f"{dates[i]}: {cloud[i]}% cloud, skip")
    i += 1

if i < len(cloud):
    print(f"Use the image of {dates[i]} ({cloud[i]}% cloud)")
else:
    print("No clear image found")
```

- `i < len(cloud)`: Makes sure we do not go past the end of the list.
- `cloud[i] >= 10`: Keep looping while the image is too cloudy.
- When the loop stops, `i` points to the first good image.

## **while True with break**
A common pattern is `while True`, which repeats forever until we stop it with `break`. It is often used for menus.

```python
# while True:
#     cmd = input("Type a city (q to quit): ")
#     if cmd == "q":
#         break
#     print("You typed", cmd.title())

# The same idea with a fixed list instead of input()
fake_inputs = ["kolkata", "siliguri", "durgapur", "q"]
i = 0
while True:
    cmd = fake_inputs[i]
    i += 1
    if cmd == "q":
        print("Bye")
        break
    print("You typed", cmd.title())
```

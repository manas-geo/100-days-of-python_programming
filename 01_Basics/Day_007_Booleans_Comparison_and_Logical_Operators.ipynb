# **Day 7: Booleans, Comparison and Logical Operators**

A **Boolean** value is either `True` or `False`. Computers use these to make decisions. Before learning `if-else`, we need to know how Python compares values.

## **Comparison Operators**
These operators compare two values and give `True` or `False`.

| Operator | Meaning |
|---|---|
| `==` | Equal to |
| `!=` | Not equal to |
| `>` | Greater than |
| `<` | Less than |
| `>=` | Greater than or equal to |
| `<=` | Less than or equal to |

```python
kolkata_temp = 34
darjeeling_temp = 14

print(kolkata_temp > darjeeling_temp)
print(kolkata_temp == darjeeling_temp)
print(kolkata_temp != darjeeling_temp)
print(darjeeling_temp <= 14)
```

```python
# Strings are compared letter by letter
print("apple" < "banana")
print("Kolkata" == "kolkata")     # Capital and small letters are different
```

## **Chained Comparison**
Python lets us write `a < b < c`, just like in maths.

```python
temp = 22
print(15 <= temp <= 25)     # Is the weather pleasant?
```

## **Logical Operators**
Logical operators join two or more conditions.

| Operator | Meaning |
|---|---|
| `and` | True only if **both** are True |
| `or` | True if **at least one** is True |
| `not` | Opposite (True becomes False, False becomes True) |

```python
is_sunday = True
is_raining = False

print(is_sunday and not is_raining)    # Can we go to the Maidan?
print(is_sunday or is_raining)
print(not is_sunday)
```

**Truth Table**

| A | B | A and B | A or B |
|---|---|---|---|
| True | True | True | True |
| True | False | False | True |
| False | True | False | True |
| False | False | False | False |

## **Example from GIS: Is the Point Inside West Bengal?**
A **bounding box** is the smallest rectangle that covers an area on the map. It is defined by four values: minimum and maximum latitude, and minimum and maximum longitude.

West Bengal's bounding box is roughly:
- Latitude: **21.5 to 27.3** degree North
- Longitude: **85.8 to 89.9** degree East

If a point's latitude **and** longitude are both inside these limits, the point is inside the box.

```python
min_lat, max_lat = 21.5, 27.3
min_lon, max_lon = 85.8, 89.9

# Siliguri
lat, lon = 26.7271, 88.3953
print("Siliguri inside box:", min_lat <= lat <= max_lat and min_lon <= lon <= max_lon)

# Patna (Bihar)
lat, lon = 25.5941, 85.1376
print("Patna inside box:", min_lat <= lat <= max_lat and min_lon <= lon <= max_lon)
```

Let us understand the code above:
- `min_lat <= lat <= max_lat`: Checks if the latitude is between the minimum and maximum.
- `and`: Both the latitude check **and** the longitude check must be True.
- Patna's longitude (85.13) is less than 85.8, so it is outside the box.

Note: A bounding box is only a rough check. Some places inside the box may still be outside the real border, because West Bengal's shape is not a rectangle.

## **Short Circuit Behaviour**
- `and` stops at the first `False` value.
- `or` stops at the first `True` value.

Also, they return the actual value, not always `True`/`False`. This is useful for setting a **default value**.

```python
print(0 and 5)
print(3 and 5)
print(0 or 5)

user_city = ""
city = user_city or "Kolkata"     # If user_city is empty, use Kolkata
print(city)
```

## **Membership and Identity Operators**
- `in` / `not in`: Checks if a value is inside a collection.
- `is` / `is not`: Checks if two names point to the **same object** in memory. It is mostly used to check `None`.

```python
districts = ["Darjeeling", "Kalimpong", "Jalpaiguri", "Alipurduar", "Cooch Behar"]
print("Kalimpong" in districts)
print("Purulia" in districts)

x = None
print(x is None)
```

# **Day 19: Dictionaries**

A **dictionary** stores data in **key: value** pairs, just like a real dictionary stores a word and its meaning.

- The **key** is the name (it must be unique).
- The **value** is the data for that key.

For example: `"pin": 734001` means the key `"pin"` has the value `734001`.

```python
city = {
    "name": "Siliguri",
    "district": "Darjeeling",
    "pin": 734001,
    "famous_for": ["tea", "gateway to the hills", "NJP station"],
}
print(city)
print(city["name"])
print(city["famous_for"][0])
```

## **Reading Values Safely with get()**
If we use a key that does not exist with `[]`, Python gives an error. The `get()` method returns `None` (or a default value) instead.

```python
print(city.get("pin"))
print(city.get("population"))
print(city.get("population", "Not available"))

# print(city["population"])      # This will raise a KeyError
```

## **Adding, Changing and Deleting**

```python
city["population"] = 513264          # Adding a new key (Census 2011, approx)
city["pin"] = 734001                 # Changing an existing key
print(city)

del city["population"]
removed = city.pop("famous_for")
print("Removed:", removed)
print(city)
```

## **Looping Through a Dictionary**
- `.keys()` gives all keys.
- `.values()` gives all values.
- `.items()` gives both key and value together.

```python
std_codes = {"Kolkata": "033", "Siliguri": "0353", "Kharagpur": "03222",
             "Durgapur": "0343", "Darjeeling": "0354"}

for k, v in std_codes.items():
    print(f"{k:<12} {v}")

print(list(std_codes.keys()))
print(list(std_codes.values()))
```

## **Example from Remote Sensing: Sentinel-2 Bands**
A dictionary is perfect for storing information about satellite bands. The **key** is the band name, and the **value** is another dictionary with details.

(Wavelength is the "colour" of light measured in nanometres (nm). Resolution is the pixel size.)

```python
sentinel2 = {
    "B2": {"name": "Blue", "wavelength_nm": 490, "resolution_m": 10},
    "B3": {"name": "Green", "wavelength_nm": 560, "resolution_m": 10},
    "B4": {"name": "Red", "wavelength_nm": 665, "resolution_m": 10},
    "B8": {"name": "NIR", "wavelength_nm": 842, "resolution_m": 10},
    "B11": {"name": "SWIR 1", "wavelength_nm": 1610, "resolution_m": 20},
    "B12": {"name": "SWIR 2", "wavelength_nm": 2190, "resolution_m": 20},
}

print("Band B8 is", sentinel2["B8"]["name"])

for band, info in sentinel2.items():
    print(f"{band:<4} {info['name']:<7} {info['wavelength_nm']:>5} nm  {info['resolution_m']} m")
```

```python
# Which bands have 10 metre resolution?
ten_m = []
for band, info in sentinel2.items():
    if info["resolution_m"] == 10:
        ten_m.append(band)
print("10 m bands:", ten_m)
```

Let us understand the code above:
- `sentinel2["B8"]["name"]`: First we get the value of `"B8"` (a dictionary), then we get `"name"` from it.
- `info['resolution_m'] == 10`: We check each band and keep only the 10 m bands. These are the ones we usually use to make a sharp colour image.

## **Counting with a Dictionary**
A very common use of a dictionary is **counting** how many times each thing appears.

```python
text = "kolkata kolkata howrah siliguri kolkata howrah digha"
freq = {}
for w in text.split():
    freq[w] = freq.get(w, 0) + 1
print(freq)
```

```python
# Count land cover classes in a classified image (a flat list of pixel labels)
pixels = ["water", "forest", "forest", "urban", "water", "forest", "agri", "agri", "forest", "urban"]
class_count = {}
for p in pixels:
    class_count[p] = class_count.get(p, 0) + 1

for cls, n in class_count.items():
    print(f"{cls:<7} {n} pixels ({n / len(pixels):.0%})")
```

- `freq.get(w, 0) + 1`: If the word is not yet in the dictionary, `get()` gives 0, so we start counting from 1.

## **Making a Dictionary from Two Lists**

```python
districts = ["Bankura", "Purulia", "Birbhum"]
headquarters = ["Bankura", "Purulia", "Suri"]

d = dict(zip(districts, headquarters))
print(d)
```

# **Day 15: Lists**

A **list** is a collection of many items kept **in order** inside square brackets `[ ]`.

**Features of a list:**
- **Ordered:** Items stay in the order we put them.
- **Changeable (mutable):** We can add, remove or change items.
- **Allows duplicates:** The same value can appear more than once.
- **Mixed types:** One list can hold numbers, text, True/False together.

## **Creating a List**

```python
districts = ["Kolkata", "Howrah", "Hooghly", "Nadia", "Murshidabad"]
print(districts)
print(type(districts))
print(len(districts))
```

```python
mixed = ["Siliguri", 734001, 26.72, True]
empty = []
print(mixed)
print(empty, len(empty))
```

## **Accessing Items (Indexing and Slicing)**
Lists use the same indexing and slicing as strings.

```python
print(districts[0])       # First item
print(districts[-1])      # Last item
print(districts[1:3])     # Items at index 1 and 2
print(districts[::-1])    # Reversed list
```

## **Changing Items**
Unlike strings, lists **can be changed**.

```python
districts[0] = "Kolkata (KMC)"
print(districts)
```

## **Looping Through a List**

```python
for d in districts:
    print(d.upper())
```

## **Joining and Repeating Lists**

```python
north = ["Darjeeling", "Jalpaiguri"]
south = ["Purba Medinipur", "Paschim Medinipur"]

print(north + south)
print(["Dhak"] * 4)
```

## **Nested List (List Inside a List)**
A nested list is like a **table**. Each inner list is one row.

```python
# Name, Bengali, English, Maths
students = [
    ["Riya", 88, 76, 92],
    ["Arjun", 72, 81, 65],
    ["Sneha", 95, 89, 98],
]

for s in students:
    total = s[1] + s[2] + s[3]
    print(f"{s[0]:<6} Total = {total}, Average = {total / 3:.1f}")

print("Sneha's Maths marks:", students[2][3])
```

## **Useful Functions for Lists**

```python
temps = [31, 34, 36, 29, 33, 38, 35]      # One week in Kolkata, April

print("Maximum:", max(temps))
print("Minimum:", min(temps))
print("Sum:", sum(temps))
print("Average:", round(sum(temps) / len(temps), 1))
print("Sorted:", sorted(temps))
print("Original is not changed:", temps)
```

## **Example from GIS: A Route as a List of Coordinates**
In GIS, a road or a river is stored as a **list of points**. Each point is a `[longitude, latitude]` pair. Joining the points in order draws the line on the map.

Below is a very simple version of the route from Kolkata to Digha.

```python
route = [
    [88.36, 22.57],    # Kolkata
    [88.10, 22.35],    # Uluberia side
    [87.92, 22.30],    # Tamluk side
    [87.78, 21.90],    # Contai side
    [87.51, 21.63],    # Digha
]

print("Number of points:", len(route))
print("Start point:", route[0])
print("End point:", route[-1])

# Separate the longitudes and latitudes
lons = []
lats = []
for point in route:
    lons.append(point[0])
    lats.append(point[1])

print("Longitudes:", lons)
print("Latitudes:", lats)
print("Bounding box -> lon:", min(lons), "to", max(lons), " lat:", min(lats), "to", max(lats))
```

Let us understand the code above:
- `route[0]` is the first point (Kolkata) and `route[-1]` is the last point (Digha).
- `point[0]` is the longitude and `point[1]` is the latitude of each point.
- Using `min()` and `max()`, we get the **bounding box** of the route (the smallest rectangle that covers it).

## **Copying a List (Important)**
If we write `b = a`, we do **not** get a new list. Both names point to the **same** list. To make a real copy, use `.copy()`.

```python
a = ["ilish", "rui"]
b = a              # Same list, two names
b.append("katla")
print("a:", a)     # a is also changed!

a = ["ilish", "rui"]
b = a.copy()       # A real copy
b.append("katla")
print("a:", a)
print("b:", b)
```

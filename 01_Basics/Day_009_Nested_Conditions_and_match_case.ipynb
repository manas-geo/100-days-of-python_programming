## **Nested if**
A **nested if** is an `if` inside another `if`. We use it when a second question depends on the answer to the first question.

```python
has_ticket = True
seat_confirmed = False
train = "Darjeeling Mail"

if has_ticket:
    if seat_confirmed:
        print("Board the", train)
    else:
        print("Ticket is in waiting list, check PNR")
else:
    print("Book a ticket first")
```

Too much nesting makes code hard to read. Often we can make it flat using `elif`:

```python
if not has_ticket:
    print("Book a ticket first")
elif seat_confirmed:
    print("Board the", train)
else:
    print("Ticket is in waiting list, check PNR")
```

## **Example: Largest of Three Numbers**

```python
a, b, c = 45, 78, 62

if a >= b and a >= c:
    print("Largest:", a)
elif b >= c:
    print("Largest:", b)
else:
    print("Largest:", c)

print("Using max():", max(a, b, c))
```

## **Example from Remote Sensing: Is This Image Useful?**
Before using a satellite image, we check two things:
1. Is the **cloud cover** low enough? (Clouds hide the ground.)
2. Is it from the right **season**?

```python
cloud_cover = 8        # In percent
month = 12             # December

if cloud_cover <= 10:
    if month in (11, 12, 1, 2):
        print("Good image: low cloud and winter (clear sky season)")
    else:
        print("Low cloud, but check the season")
else:
    print("Too cloudy, search for another date")
```

## **match-case**
`match-case` (Python 3.10 and newer) checks one value against many possible options. It is a cleaner way to write a long `if-elif` chain.
- `case "A" | "B"`: matches A **or** B.
- `case _`: matches **anything else** (like `else`).

```python
day = "sunday"

match day:
    case "saturday" | "sunday":
        print("Weekend: mangsho bhaat day")
    case "monday":
        print("Back to work")
    case _:
        print("Normal weekday")
```

## **Example from GIS: Land Cover Codes**
In a classified satellite image, every pixel stores a **number code** instead of a name. For example, `1` may mean Water and `2` may mean Forest. Let us convert the code into a name using `match-case`.

```python
pixel_code = 3

match pixel_code:
    case 1:
        name = "Water"
    case 2:
        name = "Forest"
    case 3:
        name = "Agriculture"
    case 4:
        name = "Built-up"
    case 5:
        name = "Bare land"
    case _:
        name = "Unknown"

print("Pixel code", pixel_code, "means", name)
```

## **match with Tuples**
`match` can also look inside a group of values like a coordinate pair `(x, y)`.

```python
point = (0, 5)

match point:
    case (0, 0):
        print("Origin")
    case (0, y):
        print(f"On the Y axis at {y}")
    case (x, 0):
        print(f"On the X axis at {x}")
    case (x, y):
        print(f"Point at {x}, {y}")
```

## **match with a Condition (Guard)**
We can add an `if` after a case. This is called a **guard**.

```python
metro_trip_km = 14

match metro_trip_km:
    case k if k <= 2:
        fare = 5
    case k if k <= 5:
        fare = 10
    case k if k <= 10:
        fare = 15
    case k if k <= 20:
        fare = 20
    case _:
        fare = 25

print("Approximate metro fare:", fare)
```

(These fare slabs are roughly like the Kolkata Metro, they may not be exact.)

## **Practice Problem: Type of Triangle**

```python
a, b, c = 5, 5, 8

if a + b <= c or a + c <= b or b + c <= a:
    print("Not a valid triangle")
elif a == b == c:
    print("Equilateral")
elif a == b or b == c or a == c:
    print("Isosceles")
else:
    print("Scalene")
```

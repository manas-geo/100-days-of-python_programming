# **Day 83: Pandas - Series and DataFrame**

**Pandas** is like **Excel inside Python**. It is used to work with tables of data: reading, cleaning, filtering, grouping and summarising.

Install it with `pip install pandas`.

Pandas has two main objects:
- **Series:** One column of data (with labels called the **index**).
- **DataFrame:** A full table (many Series that share the same index).

```python
import pandas as pd
import numpy as np
print(pd.__version__)
```

## **Series**

```python
population = pd.Series([4496694, 4850029, 5913457, 1846823],
                       index=["Kolkata", "Howrah", "Paschim Medinipur", "Darjeeling"],
                       name="population")
print(population)
print()
print("Howrah:", population["Howrah"])
print(population[population > 4_600_000])
```

## **DataFrame: West Bengal Districts**
The data below is from the **Census 2011**, with the 19 districts that existed before the new districts were made (2014 to 2017). Population numbers are from the census tables. Area and literacy are approximate.

```python
data = {
    "district": ["Kolkata", "Howrah", "North 24 Parganas", "South 24 Parganas", "Hooghly", "Nadia",
                 "Purba Medinipur", "Paschim Medinipur", "Bankura", "Purulia", "Bardhaman", "Birbhum",
                 "Murshidabad", "Malda", "Uttar Dinajpur", "Dakshin Dinajpur", "Jalpaiguri", "Darjeeling", "Cooch Behar"],
    "hq": ["Kolkata", "Howrah", "Barasat", "Alipore", "Chinsurah", "Krishnanagar",
           "Tamluk", "Midnapore", "Bankura", "Purulia", "Bardhaman", "Suri",
           "Baharampur", "English Bazar", "Raiganj", "Balurghat", "Jalpaiguri", "Darjeeling", "Cooch Behar"],
    "division": ["Presidency", "Presidency", "Presidency", "Presidency", "Burdwan", "Presidency",
                 "Medinipur", "Medinipur", "Medinipur", "Medinipur", "Burdwan", "Burdwan",
                 "Malda", "Malda", "Malda", "Malda", "Jalpaiguri", "Jalpaiguri", "Jalpaiguri"],
    "population": [4496694, 4850029, 10009781, 8161961, 5519145, 5167600,
                   5095875, 5913457, 3596674, 2930115, 7717563, 3502404,
                   7103807, 3988845, 3007134, 1676276, 3872846, 1846823, 2819086],
    "area_km2": [206, 1467, 4094, 9960, 3149, 3927, 4736, 9296, 6882, 6259, 7024, 4545,
                 5324, 3733, 3140, 2219, 6227, 3149, 3387],
    "literacy": [86.3, 83.3, 84.1, 77.5, 82.0, 74.9, 87.0, 78.0, 70.3, 64.5, 76.2, 70.7,
                 66.6, 61.7, 59.1, 72.8, 73.3, 79.6, 74.8],
}
df = pd.DataFrame(data)
df.head()
```

## **First Look at the Data**

```python
df.describe().round(1)         # Quick statistics of number columns
```

## **Selecting Columns and Rows**
- `df["column"]` gives one column.
- `df[["col1", "col2"]]` gives many columns.
- `df.loc[...]` selects by **label** (name).
- `df.iloc[...]` selects by **position** (number).

```python
df[["district", "literacy"]].head(3)
```

```python
df = df.set_index("district")          # Use district names as row labels
df.loc["Paschim Medinipur"]
```

```python
df.loc[["Kolkata", "Darjeeling"], ["hq", "population"]]
```

```python
df.iloc[0:3, 0:2]
```

## **Filtering Rows**

```python
df[df["literacy"] > 80]
```

```python
df[(df["division"] == "Medinipur") & (df["population"] > 4_000_000)]
```

```python
df.query("area_km2 < 3500 and literacy > 75")
```

## **Adding New Columns**
**Population density** = people per square km. It is one of the most important numbers in population geography.

```python
df["density"] = (df["population"] / df["area_km2"]).round()
df["pop_lakh"] = (df["population"] / 1e5).round(1)
df["size"] = np.where(df["area_km2"] > 5000, "Large", "Small")

df.sort_values("density", ascending=False)[["density", "pop_lakh", "size"]].head(5)
```

Kolkata's density is extremely high because the district is only the city itself (about 206 square km).

## **Quick Summaries**

```python
print(f"Total population of West Bengal: {df['population'].sum():,}")
print("Most literate :", df["literacy"].idxmax(), df["literacy"].max())
print("Least literate:", df["literacy"].idxmin(), df["literacy"].min())
print()
print(df["division"].value_counts())
```

The total, **91,276,115**, matches the official Census 2011 population of West Bengal.

## **Example from GIS: An Attribute Table**
In GIS software like QGIS, every layer has an **attribute table**: one row per feature, one column per property. A Pandas DataFrame is exactly the same idea. (The GeoPandas library, which we meet on Day 99, even adds a **geometry** column to a DataFrame.)

Let us make a table of GPS survey points and summarise it.

```python
points = pd.DataFrame({
    "point_id": ["SB001", "SB002", "SB003", "SB004", "SB006", "SB009", "SB011", "SB012"],
    "lat": [21.9452, 21.9610, 21.9877, 22.0150, 21.9720, 22.0412, 21.9901, 22.0101],
    "lon": [88.8931, 88.9105, 88.8702, 88.8450, 88.9020, 88.8390, 88.8801, 88.8599],
    "land_cover": ["Mangrove", "Water", "Mangrove", "Agriculture", "Mudflat", "Agriculture", "Mangrove", "Settlement"],
})
print(points["land_cover"].value_counts())
print("Survey area centre:", points["lat"].mean().round(4), points["lon"].mean().round(4))
```

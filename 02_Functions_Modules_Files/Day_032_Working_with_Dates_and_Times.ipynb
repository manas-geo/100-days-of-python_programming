# **Day 32: Working with Dates and Times**

The `datetime` module helps us work with dates and times. It has four main parts:

| Part | Stores | Example |
|---|---|---|
| `date` | Only a date | 2024-10-10 |
| `time` | Only a time | 18:45 |
| `datetime` | Date and time together | 2024-10-10 18:45 |
| `timedelta` | A length of time | 5 days, 3 hours |

```python
from datetime import date, time, datetime, timedelta

d = date(2024, 10, 10)
print(d, d.year, d.month, d.day)
print(d.weekday())          # Monday = 0, Sunday = 6
print(d.strftime("%A"))     # Name of the day
```

## **Formatting Dates with strftime()**
`strftime()` changes a date into text in any format we want.

| Code | Meaning | Example |
|---|---|---|
| `%d` | Day | 10 |
| `%m` | Month number | 10 |
| `%Y` | Year | 2024 |
| `%B` | Month name | October |
| `%b` | Short month name | Oct |
| `%A` | Day name | Thursday |
| `%H:%M` | 24-hour time | 18:45 |
| `%I:%M %p` | 12-hour time | 06:45 PM |
| `%j` | Day of the year | 284 |

```python
dt = datetime(2024, 10, 10, 18, 45)
print(dt.strftime("%d/%m/%Y"))
print(dt.strftime("%d %B %Y, %A"))
print(dt.strftime("%I:%M %p"))
```

## **Reading Dates from Text with strptime()**
`strptime()` does the opposite: it changes text into a date.

```python
s = "25-12-2024 21:30"
x = datetime.strptime(s, "%d-%m-%Y %H:%M")
print(x, type(x))

print(date.fromisoformat("2024-04-14"))     # Poila Baisakh
```

## **Date Calculations with timedelta**

```python
puja_start = date(2024, 10, 9)
print("Dashami:", puja_start + timedelta(days=4))

exam = date(2025, 2, 10)
today = date(2024, 11, 1)
left = exam - today
print(left.days, "days left for the exam")
```

```python
# Train journey
departure = datetime(2024, 5, 20, 22, 5)
travel = timedelta(hours=10, minutes=15)
arrival = departure + travel
print("Arrives at NJP:", arrival.strftime("%d %b %I:%M %p"))
```

## **Example from Remote Sensing: Satellite Revisit Dates**
A satellite passes over the same place again after a fixed number of days. This is called the **revisit time**.
- **Sentinel-2** (two satellites together): every **5 days**
- **Landsat 8**: every **16 days**

If we know one date when Sentinel-2 took an image of Kolkata, we can find the next dates.

```python
first_pass = date(2024, 1, 5)
revisit = timedelta(days=5)

print("Next Sentinel-2 dates over Kolkata:")
d = first_pass
for _ in range(6):
    print(" ", d.strftime("%d %b %Y (%a)"))
    d += revisit
```

## **Day of Year (Julian Day)**
Many satellite products use the **day of the year** (1 to 365 or 366) instead of the normal date. For example, MODIS files have names like `A2024045`, which means **year 2024, day 45**.

```python
# Date -> day of year
d = date(2024, 2, 14)
print(d.strftime("%j"), d.timetuple().tm_yday)

# Day of year -> date
code = "A2024045"
year = int(code[1:5])
doy = int(code[5:8])
real_date = date(year, 1, 1) + timedelta(days=doy - 1)
print(code, "means", real_date.strftime("%d %B %Y"))
```

- `date(year, 1, 1) + timedelta(days=doy - 1)`: Start from 1st January and add (day number - 1) days.

## **Age Calculator**

```python
def age_on(dob, on_date):
    years = on_date.year - dob.year
    if (on_date.month, on_date.day) < (dob.month, dob.day):
        years -= 1           # Birthday has not come yet this year
    return years

print(age_on(date(2004, 8, 15), date(2024, 8, 14)))
print(age_on(date(2004, 8, 15), date(2024, 8, 15)))
```

## **Sorting Festival Dates**

```python
events = {
    "Rath Yatra": date(2024, 7, 7),
    "Durga Puja (Shashthi)": date(2024, 10, 9),
    "Poila Baisakh": date(2024, 4, 14),
    "Kali Puja": date(2024, 10, 31),
    "Kolkata Book Fair": date(2024, 1, 18),
}
for name, d in sorted(events.items(), key=lambda kv: kv[1]):
    print(f"{d:%d %b}  {name}")
```

## **Practice Problem: How Many Sundays in 2024?**

```python
d = date(2024, 1, 1)
sundays = 0
while d.year == 2024:
    if d.weekday() == 6:
        sundays += 1
    d += timedelta(days=1)
print(sundays)
```

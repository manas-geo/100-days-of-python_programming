# **Day 21: Functions**

A **function** is a named block of code that does one job. We write it once and can use it (call it) again and again.

Think of a function like a **mixer grinder**: you put something in (input), it does its work, and you get something out (output). You don't need to know what happens inside every time you use it.

**Why use functions?**
- **Reuse:** Write once, use many times.
- **Clean code:** A big program is broken into small, easy parts.
- **Easy to fix:** If there is a mistake, we fix it in one place only.

## **Creating and Calling a Function**
- We create a function with the `def` keyword.
- We call it by writing its name with brackets `()`.

```python
def greet():
    print("Nomoskar!")

greet()
greet()
```

## **Parameters and Arguments**
- A **parameter** is the name written inside the brackets when **creating** the function.
- An **argument** is the actual value we give when **calling** the function.

```python
def greet(name, city):          # name and city are parameters
    print(f"Nomoskar {name}, welcome to {city}")

greet("Rahul", "Siliguri")      # "Rahul" and "Siliguri" are arguments
greet("Priya", "Midnapore")

# greet("Rahul")                # This will raise a TypeError (city is missing)
```

## **The return Statement**
`return` sends a value **back** from the function, so that we can store it or use it later.

**print vs return:**
- `print()` only **shows** the value on the screen.
- `return` **gives back** the value to the place where the function was called.

```python
def area_of_rectangle(length, breadth):
    return length * breadth

a = area_of_rectangle(20, 15)
print("Plot area:", a, "square feet")
print("Double the area:", a * 2)
```

```python
# A function without return gives back None
def area_print(length, breadth):
    print(length * breadth)

x = area_print(20, 15)
print("x is", x)
```

## **Docstring**
A **docstring** is the first text inside a function (in triple quotes). It explains what the function does. `help()` shows it.

```python
def bmi(weight_kg, height_m):
    """Return the Body Mass Index rounded to 1 decimal place."""
    return round(weight_kg / height_m ** 2, 1)

print(bmi(68, 1.72))
help(bmi)
```

## **Example from Remote Sensing: The NDVI Function**
Now we can finally calculate **NDVI**. The formula is:

**NDVI = (NIR - Red) / (NIR + Red)**

- **NIR** is the Near Infrared band value.
- **Red** is the Red band value.

Healthy plants reflect a lot of NIR and very little red light, so their NDVI is high (close to +1). Water reflects almost no NIR, so its NDVI is negative.

```python
def ndvi(nir, red):
    """Calculate NDVI from NIR and Red reflectance values."""
    if nir + red == 0:
        return 0
    return (nir - red) / (nir + red)

print("Forest :", round(ndvi(0.45, 0.05), 3))
print("Crop   :", round(ndvi(0.30, 0.10), 3))
print("Soil   :", round(ndvi(0.25, 0.20), 3))
print("Water  :", round(ndvi(0.02, 0.06), 3))
```

Let us understand the code above:
- `if nir + red == 0`: We cannot divide by zero, so we return 0 in that case.
- For the forest pixel: (0.45 - 0.05) / (0.45 + 0.05) = 0.40 / 0.50 = **0.8**.
- For water, NIR is smaller than red, so the NDVI is **negative**.

## **A Function Calling Another Function**

```python
def classify_ndvi(value):
    """Convert an NDVI value into a land cover name."""
    if value < 0:
        return "Water"
    elif value < 0.2:
        return "Bare / Built-up"
    elif value < 0.5:
        return "Sparse vegetation"
    return "Dense vegetation"

def describe_pixel(nir, red):
    v = ndvi(nir, red)
    return f"NDVI {v:.2f} -> {classify_ndvi(v)}"

print(describe_pixel(0.45, 0.05))
print(describe_pixel(0.02, 0.06))
```

## **Functions Are Objects**
In Python, a function is also an object. We can store it in a variable or even inside a dictionary.

```python
def add(a, b):
    return a + b

def sub(a, b):
    return a - b

operations = {"+": add, "-": sub}
print(operations["+"](10, 5))
print(operations["-"](10, 5))
```

## **Practice Problems**
1. Write a function that returns the grade for given marks.
2. Write a function that checks if a PIN code belongs to West Bengal (WB PIN codes start with 70, 71, 72, 73 or 74).

```python
def grade(marks):
    if marks >= 90: return "AA"
    if marks >= 80: return "A+"
    if marks >= 60: return "A"
    if marks >= 45: return "B"
    if marks >= 35: return "C"
    return "Fail"

for m in [95, 83, 61, 40, 20]:
    print(m, grade(m))
```

```python
def is_wb_pin(pin):
    s = str(pin)
    return len(s) == 6 and s[:2] in ("70", "71", "72", "73", "74")

for p in [700001, 721101, 734001, 110001]:
    print(p, is_wb_pin(p))
```

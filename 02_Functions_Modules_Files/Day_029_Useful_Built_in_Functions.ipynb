# **Day 29: Useful Built-in Functions**

Python has about 70 **built-in functions** that are always ready to use, without importing anything. Here are the ones we use most often.

```python
import builtins
names = [n for n in dir(builtins) if not n.startswith("_") and n[0].islower()]
print(len(names))
print(names)
```

## **zip() and enumerate()**

```python
districts = ["Kolkata", "Howrah", "Hooghly"]
headquarters = ["Kolkata", "Howrah", "Chinsurah"]

for i, (d, h) in enumerate(zip(districts, headquarters), 1):
    print(i, d, "->", h)
```

```python
# zip() stops at the shortest list
print(list(zip([1, 2, 3], "ab")))

# Unzip with zip(*)
pairs = [("a", 1), ("b", 2), ("c", 3)]
letters, numbers = zip(*pairs)
print(letters, numbers)
```

## **sorted(), min() and max() with key**
The `key` tells these functions **what to compare**.

```python
players = [("Bappa", 1363), ("Tublu", 924), ("Rana", 3000), ("Pintu", 600)]   # Local club runs

print(sorted(players, key=lambda p: p[1], reverse=True))
print(max(players, key=lambda p: p[1]))
print(min(players, key=lambda p: len(p[0])))
```

## **any() and all()**
- `any()`: True if **at least one** item is True.
- `all()`: True if **every** item is True.

```python
aqi = [110, 95, 180, 72]
print(any(a > 150 for a in aqi))    # Was any day unhealthy?
print(all(a < 200 for a in aqi))    # Were all days below 200?
```

## **Example from Remote Sensing: Bit Flags in a Quality Band**
Satellite products often come with a **quality band** (called QA band). In this band, each pixel's number hides many Yes/No answers inside its **binary bits**. For example:
- Bit 3 = 1 means **cloud**
- Bit 4 = 1 means **cloud shadow**

(This is a simplified version of how Landsat QA bands work.)

```python
def is_cloud(qa_value):
    return (qa_value >> 3) & 1 == 1          # Look at bit number 3

def is_shadow(qa_value):
    return (qa_value >> 4) & 1 == 1          # Look at bit number 4

for qa in [0, 8, 16, 24, 1]:
    print(f"QA {qa:>2} = {bin(qa):>7}  cloud={is_cloud(qa)}  shadow={is_shadow(qa)}")
```

Let us understand the code above:
- `bin(8)` is `0b1000`. Counting from the right starting at 0, bit number 3 is `1`, so it is a cloud.
- `qa_value >> 3` shifts the bits 3 places to the right, and `& 1` keeps only the last bit.
- 24 is `0b11000`, so both bit 3 and bit 4 are 1: cloud **and** shadow.

## **eval() (Use with Care)**
`eval()` runs a string as Python code. Never use it on text typed by a user, because it can run anything.

```python
print(eval("2 + 3 * 4"))
```

## **Practice Problem: Top 3 Districts by Literacy**
(Approximate Census 2011 figures.)

```python
literacy = {"Purba Medinipur": 87.0, "Kolkata": 86.3, "North 24 Parganas": 84.1,
            "Howrah": 83.3, "Hooghly": 82.0, "Uttar Dinajpur": 59.1, "Malda": 61.7}

top3 = sorted(literacy.items(), key=lambda kv: kv[1], reverse=True)[:3]
for rank, (d, p) in enumerate(top3, 1):
    print(f"{rank}. {d} {p}%")
```

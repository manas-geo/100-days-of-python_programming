# **Day 25: List Comprehension**

A **list comprehension** is a short and clean way to make a new list from an old one, in **one line**.

**Format:** `[expression for item in list if condition]`

Read it like English: "Give me `expression` for each `item` in the `list`, only `if` the condition is true."

```python
# The normal way with a loop
squares = []
for i in range(1, 11):
    squares.append(i * i)
print(squares)

# The same thing with a list comprehension
squares = [i * i for i in range(1, 11)]
print(squares)
```

```python
cities = ["kolkata", "howrah", "siliguri", "durgapur"]
print([c.title() for c in cities])
print([len(c) for c in cities])
```

## **With a Condition (Filtering)**

```python
print([c for c in cities if "pur" in c])
print([n for n in range(1, 30) if n % 3 == 0 and n % 2 == 0])
```

## **With if-else (Changing Every Item)**
**Note the position:** A plain `if` (filter) goes at the **end**. An `if-else` (change) goes at the **start**, before `for`.

```python
temps = [31, 38, 22, 40, 17]
labels = ["Hot" if t >= 35 else "OK" if t >= 20 else "Cold" for t in temps]
print(labels)
```

## **Example from Remote Sensing: NDVI for Many Pixels**
We have the Red and NIR values of many pixels. Let us calculate NDVI for all of them in one line, then classify them.

```python
red = [0.05, 0.10, 0.20, 0.06, 0.04, 0.15]
nir = [0.45, 0.30, 0.25, 0.02, 0.50, 0.35]

ndvi = [round((n - r) / (n + r), 2) for n, r in zip(nir, red)]
print("NDVI:", ndvi)

vegetation = [v for v in ndvi if v > 0.5]
print("Dense vegetation pixels:", vegetation)

classes = ["Water" if v < 0 else "Bare" if v < 0.2 else "Sparse" if v < 0.5 else "Dense" for v in ndvi]
print("Classes:", classes)
```

Let us understand the code above:
- `zip(nir, red)` gives us one NIR value and one Red value together for each pixel.
- `[v for v in ndvi if v > 0.5]` keeps only the pixels with high NDVI.
- The last line uses `if-else` inside the comprehension to give a class name to every pixel.

## **Nested Comprehension (Two Loops)**

```python
pairs = [(a, b) for a in "AB" for b in range(1, 4)]
print(pairs)

# Make a flat list from a 2D list (like turning an image grid into one long list)
grid = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flat = [x for row in grid for x in row]
print(flat)

# Transpose (rows become columns)
print([[row[i] for row in grid] for i in range(3)])
```

## **Example: Trains Leaving After 8 PM to North Bengal**

```python
trains = [
    {"name": "Darjeeling Mail", "to": "NJP", "dep": "22:05"},
    {"name": "Padatik Express", "to": "NJP", "dep": "23:20"},
    {"name": "Kanchankanya Express", "to": "Alipurduar", "dep": "20:35"},
    {"name": "Howrah Puri Express", "to": "Puri", "dep": "22:35"},
    {"name": "Vande Bharat", "to": "NJP", "dep": "05:55"},
]
north = {"NJP", "Alipurduar", "New Cooch Behar"}

night_trains = [t["name"] for t in trains if t["to"] in north and t["dep"] >= "20:00"]
print(night_trains)
```

(The train timings are only examples.)

## **When Not to Use a List Comprehension**
- When it becomes too long or hard to read. A normal loop is better then.
- When you only want to print something (use a normal loop for that).

# **Day 24: Lambda, map, filter and reduce**

## **Lambda Function**
A **lambda** is a small function written in **one line** and without a name.

**Format:** `lambda inputs: result`

```python
square = lambda x: x * x
print(square(7))

add = lambda a, b: a + b
print(add(3, 4))
```

Lambdas are mostly used as a short function **inside** other functions like `sorted()`, `max()`, `map()` and `filter()`.

```python
# District and area in square km (approximate)
districts = [("Kolkata", 206), ("Howrah", 1467), ("Darjeeling", 3149), ("Purulia", 6259), ("Bankura", 6882)]

print(sorted(districts, key=lambda d: d[1]))      # Sort by area
print(max(districts, key=lambda d: d[1]))         # Largest area
```

## **map(): Apply a Function to Every Item**
`map(function, list)` runs the function on each item and gives the results.

```python
prices_usd = [2, 5, 10]
prices_inr = list(map(lambda p: p * 83, prices_usd))
print(prices_inr)

nums = list(map(int, "10 20 30".split()))
print(nums)
```

## **filter(): Keep Only Some Items**
`filter(function, list)` keeps only the items for which the function returns `True`.

```python
temps = [31, 38, 29, 40, 35, 27, 41]
heatwave_days = list(filter(lambda t: t >= 38, temps))
print(heatwave_days)
```

## **reduce(): Combine All Items into One Value**
`reduce()` takes two items, combines them, then combines the result with the next item, and so on. It is in the `functools` module.

```python
from functools import reduce

nums = [1, 2, 3, 4, 5]
print(reduce(lambda a, b: a * b, nums))     # 1 x 2 x 3 x 4 x 5
```

```python
# See how reduce works step by step
def show_add(a, b):
    print(f"{a} + {b} = {a + b}")
    return a + b

reduce(show_add, [10, 20, 30, 40])
```

## **Example from Remote Sensing: Processing a List of Scenes**
Each scene is a dictionary. Let us use `filter` to keep good scenes, `map` to pull out the dates, and `reduce` to find the clearest one.

```python
scenes = [
    {"id": "S2A_20240105", "cloud": 12, "tile": "45QXF"},
    {"id": "S2B_20240110", "cloud": 3, "tile": "45QXF"},
    {"id": "S2A_20240115", "cloud": 48, "tile": "45QXF"},
    {"id": "S2B_20240120", "cloud": 0, "tile": "45QXF"},
    {"id": "S2A_20240125", "cloud": 25, "tile": "45QXF"},
]

good = list(filter(lambda s: s["cloud"] <= 15, scenes))
print("Good scenes:", len(good))

dates = list(map(lambda s: s["id"][-8:], good))
print("Their dates:", dates)

clearest = reduce(lambda a, b: a if a["cloud"] <= b["cloud"] else b, scenes)
print("Clearest scene:", clearest["id"], "with", clearest["cloud"], "% cloud")
```

- `s["id"][-8:]`: The last 8 characters of the ID are the date (YYYYMMDD).
- In `reduce`, we compare two scenes each time and keep the one with less cloud.

## **map/filter vs List Comprehension**
The same thing can often be written more simply with a **list comprehension**, which we learn tomorrow.

```python
print(list(map(lambda x: x ** 2, range(6))))
print([x ** 2 for x in range(6)])

print(list(filter(lambda x: x % 2 == 0, range(10))))
print([x for x in range(10) if x % 2 == 0])
```

## **Practice Problem: Clean a Messy List of Names**

```python
raw = ["  rahul ", "PRIYA", "sOuRaV  ", " ankita"]

clean = list(map(lambda s: s.strip().title(), raw))
print(clean)

long_names = list(filter(lambda s: len(s) > 5, clean))
print(long_names)
```

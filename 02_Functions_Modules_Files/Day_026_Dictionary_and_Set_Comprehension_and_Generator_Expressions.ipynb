# **Day 26: Dictionary and Set Comprehension, and Generator Expressions**

Yesterday we made lists in one line. The same idea works for **dictionaries** and **sets** too.

## **Dictionary Comprehension**
**Format:** `{key: value for item in list}`

```python
cities = ["Kolkata", "Howrah", "Siliguri", "Asansol"]
print({c: len(c) for c in cities})

squares = {n: n * n for n in range(1, 6)}
print(squares)
```

```python
# Keep only cities with population of 5 lakh or more (Census 2011, rounded)
pop_lakh = {"Kolkata": 44.9, "Howrah": 10.7, "Durgapur": 5.2, "Asansol": 5.6, "Siliguri": 5.1, "Bardhaman": 3.1}
big = {k: v for k, v in pop_lakh.items() if v >= 5}
print(big)
```

```python
# Swap keys and values
code = {"HWH": "Howrah", "SDAH": "Sealdah", "KGP": "Kharagpur", "NJP": "New Jalpaiguri"}
print({v: k for k, v in code.items()})
```

## **Example from Remote Sensing: Band Wavelengths**
Let us make a dictionary of band names and their wavelength in **micrometre** (µm) from a dictionary in **nanometre** (nm). (1 µm = 1000 nm.)

```python
wavelength_nm = {"Blue": 490, "Green": 560, "Red": 665, "NIR": 842, "SWIR1": 1610, "SWIR2": 2190}

wavelength_um = {band: nm / 1000 for band, nm in wavelength_nm.items()}
print(wavelength_um)

# Only the bands that humans cannot see (visible light is about 400 to 700 nm)
invisible = {band: nm for band, nm in wavelength_nm.items() if nm > 700}
print("Invisible to our eyes:", invisible)
```

Satellites can "see" light that our eyes cannot, like NIR and SWIR. This is why satellite images can show things like plant health and soil moisture.

## **Set Comprehension**
**Format:** `{item for item in list}`. The result has no duplicates.

```python
words = "ami tumi ami se tumi amra".split()
print({w for w in words})
print({len(w) for w in words})
```

## **Generator Expression**
A **generator expression** looks like a list comprehension but uses round brackets `( )`.

**The difference:** A list comprehension makes **all** the values at once and stores them in memory. A generator makes values **one at a time**, only when needed. This saves a lot of memory for big data.

Think of it like a **tea stall**: a list is like making 1000 cups of tea in advance. A generator is like making one cup whenever a customer asks.

```python
gen = (x * x for x in range(5))
print(gen)                  # A generator object, not the values
print(next(gen), next(gen)) # Get values one by one
print(list(gen))            # The remaining values
print(list(gen))            # Empty now, because a generator can be used only once
```

```python
import sys
big_list = [x for x in range(1_000_000)]
big_gen = (x for x in range(1_000_000))

print(sys.getsizeof(big_list), "bytes for the list")
print(sys.getsizeof(big_gen), "bytes for the generator")
```

```python
# Generators work very well with sum(), max(), any(), all()
print(sum(x * x for x in range(1, 101)))
print(any(v > 40 for v in pop_lakh.values()))
print(all(len(c) > 3 for c in cities))
```

## **Practice Problem: Marks to Grades**

```python
marks = {"Riya": 91, "Arjun": 67, "Tanmoy": 43, "Sneha": 88}
grades = {name: ("A" if m >= 80 else "B" if m >= 60 else "C") for name, m in marks.items()}
print(grades)
```

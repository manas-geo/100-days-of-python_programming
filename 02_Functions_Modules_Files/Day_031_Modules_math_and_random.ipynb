## **What is a Module?**
A **module** is simply a Python file that contains ready-made code (functions, variables). We can **import** it and use that code in our program.

Python comes with hundreds of modules already installed. This collection is called the **Standard Library**. People often say Python comes with "batteries included".

## **Different Ways to Import**

```python
import math                       # Import the whole module
print(math.sqrt(81))

import math as m                  # Import with a short name
print(m.pi)

from math import factorial, gcd   # Import only some functions
print(factorial(6), gcd(48, 180))

from random import randint as ri  # Import a function with a new name
print(ri(1, 6))
```

`from math import *` also works, but it is better to avoid it, because then we cannot easily tell where a function came from.

## **The math Module**

```python
print(math.ceil(2.1), math.floor(2.9))
print(math.log(100, 10), math.log2(1024), math.log10(1000))
print(math.hypot(3, 4))                 # Length of the diagonal of a 3 x 4 rectangle
print(math.isclose(0.1 + 0.2, 0.3))     # Safe way to compare decimals
print(math.radians(180), math.degrees(math.pi))
```

## **Example from GIS: Distance Between Two Places (Haversine Formula)**
The Earth is round, so we cannot use a simple ruler formula to find the distance between two places. The **Haversine formula** gives the **shortest distance over the Earth's surface** (called the great-circle distance) using latitude and longitude.

You do not need to remember the formula. Just understand the idea:
1. Convert the degrees into **radians** (the unit that maths functions use).
2. Use the difference in latitude and longitude.
3. Multiply by the Earth's radius (about **6371 km**).

```python
def haversine(lat1, lon1, lat2, lon2):
    """Straight line distance over the Earth's surface in km."""
    R = 6371
    p1, p2 = math.radians(lat1), math.radians(lat2)
    dp = math.radians(lat2 - lat1)
    dl = math.radians(lon2 - lon1)
    a = math.sin(dp / 2) ** 2 + math.cos(p1) * math.cos(p2) * math.sin(dl / 2) ** 2
    return 2 * R * math.asin(math.sqrt(a))

places = {
    "Kolkata": (22.5726, 88.3639),
    "Midnapore": (22.4257, 87.3199),
    "Siliguri": (26.7271, 88.3953),
    "Darjeeling": (27.0410, 88.2663),
    "Digha": (21.6266, 87.5074),
}

k = places["Kolkata"]
for name, (lat, lon) in places.items():
    if name != "Kolkata":
        print(f"Kolkata -> {name:<11}{haversine(*k, lat, lon):7.1f} km")
```

- This is the **straight line** distance (like a bird flying). The road distance is always longer. For example, Kolkata to Siliguri is about 460 km straight, but about 560 km by road.
- `*k` spreads the tuple `(22.5726, 88.3639)` into two separate values.

## **The random Module**
The `random` module is used to pick random numbers or items.

```python
import random

print(random.random())                 # Random decimal from 0 to 1
print(random.randint(1, 6))            # Like rolling a dice (1 and 6 are both included)
print(random.uniform(20, 40))          # Random decimal between 20 and 40
print(random.choice(["rosogolla", "sandesh", "pantua"]))
```

```python
players = ["Arko", "Bubai", "Chiku", "Dipu", "Esha", "Fulu", "Gopal", "Hena"]
random.shuffle(players)                # Mix the list
print(players)
print(random.sample(players, 3))       # Pick 3 different players
print(random.choices(["H", "T"], k=10))  # 10 coin tosses (repeats allowed)
```

## **seed(): Getting the Same "Random" Numbers Every Time**
Computers make random numbers using a formula. If we fix the starting point (the **seed**), we get the same numbers every time. This is useful for testing and for sharing results.

```python
random.seed(42)
print([random.randint(1, 100) for _ in range(5)])

random.seed(42)
print([random.randint(1, 100) for _ in range(5)])      # Same numbers again
```

## **Example from Remote Sensing: Random Sample Points for Field Survey**
To check the accuracy of a land cover map, GIS people visit some **random points** on the ground and see what is really there. This is called **ground truthing** or **accuracy assessment**.

Let us create 5 random points inside a box around Midnapore town.

```python
random.seed(10)

min_lat, max_lat = 22.38, 22.46
min_lon, max_lon = 87.28, 87.36

sample_points = []
for i in range(1, 6):
    lat = round(random.uniform(min_lat, max_lat), 5)
    lon = round(random.uniform(min_lon, max_lon), 5)
    sample_points.append((f"P{i}", lat, lon))

for pid, lat, lon in sample_points:
    print(pid, lat, lon)
```

## **Secure Random Numbers: secrets**
For passwords and OTPs, we should use the `secrets` module, because `random` is **not** safe for security.

```python
import string, secrets

otp = "".join(secrets.choice(string.digits) for _ in range(6))
print("OTP:", otp)

password = "".join(secrets.choice(string.ascii_letters + string.digits + "@#$") for _ in range(12))
print("Password:", password)
```

## **Practice Problem: Dice Simulation**
Roll two dice 10,000 times. Which total comes most often?

```python
random.seed(0)
counts = {}
for _ in range(10000):
    total = random.randint(1, 6) + random.randint(1, 6)
    counts[total] = counts.get(total, 0) + 1

for total in sorted(counts):
    print(f"{total:2} {'#' * (counts[total] // 50)}")
```

The total **7** comes most often, because there are the most ways to make 7 (1+6, 2+5, 3+4, 4+3, 5+2, 6+1).

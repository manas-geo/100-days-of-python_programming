# **Day 22: Function Arguments in Detail**

There are different ways to give values to a function. Let us learn them one by one.

| Type | Example | Meaning |
|---|---|---|
| Positional | `f(1, 2)` | Values are matched by their **order** |
| Keyword | `f(a=1, b=2)` | Values are matched by **name** |
| Default | `def f(a, b=5)` | If no value is given, the default is used |
| `*args` | `def f(*args)` | Takes **any number** of values |
| `**kwargs` | `def f(**kwargs)` | Takes **any number** of named values |

## **Positional and Keyword Arguments**

```python
def ticket(src, dst, cls):
    print(f"{src} -> {dst} [{cls}]")

ticket("Howrah", "NJP", "3A")                      # Positional (order matters)
ticket(dst="NJP", cls="SL", src="Sealdah")         # Keyword (order does not matter)
ticket("Howrah", cls="2A", dst="Digha")            # Mixed (positional must come first)
```

## **Default Arguments**

```python
def ticket(src, dst, cls="SL", qty=1):
    print(f"{qty} x {src} -> {dst} [{cls}]")

ticket("Howrah", "Bardhaman")
ticket("Howrah", "Bardhaman", qty=3)
```

## **A Common Mistake with Default Lists**
A default value is created **only once**, when the function is made, not every time it is called. So using a list as a default value can give strange results.

```python
def add_item(item, bag=[]):     # Not a good idea
    bag.append(item)
    return bag

print(add_item("aloo"))
print(add_item("peyaj"))        # "aloo" is still there!
```

```python
# The correct way: use None and make a new list inside
def add_item(item, bag=None):
    if bag is None:
        bag = []
    bag.append(item)
    return bag

print(add_item("aloo"))
print(add_item("peyaj"))
```

## ***args: Any Number of Values**
`*args` collects all extra values into a **tuple**.

```python
def total_bill(*prices):
    print("Received:", prices)
    return sum(prices)

print(total_bill(20, 30))
print(total_bill(15, 15, 25, 40, 10))
```

## ****kwargs: Any Number of Named Values**
`**kwargs` collects all extra named values into a **dictionary**.

```python
def profile(**info):
    for key, value in info.items():
        print(f"{key:>10}: {value}")

profile(name="Sourav", city="Behala", sport="Cricket")
```

```python
# All types together (the order must be: normal, *args, keyword, **kwargs)
def order(shop, *items, delivery=False, **extra):
    print("Shop:", shop)
    print("Items:", items)
    print("Delivery:", delivery)
    print("Extra:", extra)

order("Arsalan", "biryani", "chaap", delivery=True, note="less spicy")
```

## **Example from GIS: Area Conversion Function**
In West Bengal, land is often measured in **katha** and **bigha**, while GIS software gives area in **square metre**. Let us make a flexible function to convert between units.

- 1 katha = 720 square feet (in West Bengal)
- 1 bigha = 20 katha
- 1 hectare = 10,000 square metre
- 1 square metre = 10.7639 square feet

```python
def convert_area(value, unit="m2", to="hectare"):
    """Convert an area value from one unit to another."""
    to_m2 = {
        "m2": 1,
        "hectare": 10_000,
        "km2": 1_000_000,
        "acre": 4046.86,
        "sqft": 1 / 10.7639,
        "katha": 720 / 10.7639,
        "bigha": 20 * 720 / 10.7639,
    }
    m2 = value * to_m2[unit]
    return m2 / to_m2[to]

print(round(convert_area(90000), 2), "hectare")                           # 100 Landsat pixels
print(round(convert_area(90000, to="bigha"), 2), "bigha")
print(round(convert_area(3, unit="katha", to="sqft"), 1), "square feet")
print(round(convert_area(1, unit="bigha", to="m2"), 1), "square metre")
```

- `unit="m2"` and `to="hectare"` are **default arguments**. If we give only the value, it converts square metre to hectare.
- The dictionary `to_m2` tells how many square metres are in one unit. We first change everything to square metre, then to the unit we want.

## **Unpacking Arguments When Calling**
`*` and `**` can also be used when **calling** a function to spread out a tuple or a dictionary.

```python
def distance_text(a, b, km):
    return f"{a} to {b} is about {km} km"

trip = ("Kolkata", "Shantiniketan", 160)
print(distance_text(*trip))

trip_dict = {"a": "Kolkata", "b": "Bishnupur", "km": 150}
print(distance_text(**trip_dict))
```

## **Practice Problem: Average of Any Number of Values**

```python
def average(*nums, digits=2):
    if not nums:
        return 0
    return round(sum(nums) / len(nums), digits)

print(average(10, 20, 33))
print(average(10, 20, 33, digits=0))
print(average())
```

# **Day 28: Strings in Depth (with Bengali Text)**

In Python 3, all strings are **Unicode**. Unicode is a system that gives a number to every character of every language. So Bengali, Hindi and English all work without any extra setup.

```python
s = "কলকাতা"
print(s)
print(len(s))
print(list(s))
```

The length is 6, not 4. Why? A Bengali letter like "তা" is made of **two** Unicode characters: the letter "ত" and the vowel sign "া".

## **ord() and chr()**
- `ord()` gives the Unicode **number** of a character.
- `chr()` gives the **character** for a number.

```python
print(ord("A"), ord("a"), ord("অ"))
print(chr(2437), chr(65))

# Bengali digits start at number 0x09E6
print([chr(0x09E6 + i) for i in range(10)])
```

```python
def to_bengali_digits(n):
    return "".join(chr(0x09E6 + int(d)) for d in str(n))

print(to_bengali_digits(2024), to_bengali_digits(721101))
```

## **Encoding and Decoding**
Computers store text as **bytes**. **UTF-8** is the standard way to change text into bytes (encode) and back (decode).

```python
b = "Kolkata".encode("utf-8")
print(b, len(b))

b2 = "কলকাতা".encode("utf-8")
print(b2, len(b2))            # Each Bengali character takes 3 bytes
print(b2.decode("utf-8"))
```

## **translate(): Replace Many Characters at Once**

```python
table = str.maketrans("aeiou", "*****")
print("siliguri darjeeling".translate(table))

remove_punctuation = str.maketrans("", "", ",.!?")
print("hey! kemon acho? bhalo, tumi?".translate(remove_punctuation))
```

## **Raw Strings**
A **raw string** starts with `r`. Backslashes inside it are treated as normal characters. This is very useful for **Windows file paths**, which GIS users type all the time.

```python
print("C:\new_data\test.tif")      # \n and \t are treated as new line and tab!
print(r"C:\new_data\test.tif")     # Raw string keeps the path correct
```

## **Example from GIS: Reading a Coordinate Written as Text**
GPS devices and old maps often write coordinates as text like `22°34'21"N`. Let us convert this text into a decimal number.

```python
def parse_dms(text):
    """Convert text like 22°34'21"N into decimal degrees."""
    text = text.strip()
    direction = text[-1]                         # N, S, E or W
    text = text[:-1]
    for symbol in "°'\"":
        text = text.replace(symbol, " ")        # Replace the symbols with spaces
    d, m, s = [float(x) for x in text.split()]
    dd = d + m / 60 + s / 3600
    return round(-dd if direction in "SW" else dd, 5)

print(parse_dms("22°34'21\"N"))
print(parse_dms("88°21'50\"E"))
print(parse_dms("27°02'28\"N"))    # Darjeeling
```

Let us understand the code above:
- `text[-1]` is the last character, which tells the direction.
- We replace `°`, `'` and `"` with spaces, so `split()` gives us three numbers.
- Then we use the same formula as Day 23.

## **join() is Faster Than + for Building Big Strings**

```python
import time

t = time.time()
s = ""
for i in range(200000):
    s += "a"
t1 = time.time() - t

t = time.time()
s = "".join("a" for _ in range(200000))
t2 = time.time() - t

print(f"Using +=   : {t1:.4f} seconds")
print(f"Using join : {t2:.4f} seconds")
```

## **Practice Problems: Anagram and Secret Code**

```python
def is_anagram(a, b):
    clean = lambda s: sorted(s.lower().replace(" ", ""))
    return clean(a) == clean(b)

print(is_anagram("listen", "silent"), is_anagram("digha", "diga"))
```

```python
def caesar(text, shift):
    out = []
    for ch in text:
        if ch.isalpha() and ch.isascii():
            base = ord("A") if ch.isupper() else ord("a")
            out.append(chr((ord(ch) - base + shift) % 26 + base))
        else:
            out.append(ch)
    return "".join(out)

secret = caesar("Meet me at Esplanade", 3)
print(secret)
print(caesar(secret, -3))
```

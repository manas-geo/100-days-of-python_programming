## **What is Scope?**
**Scope** means **where a variable can be seen and used**.

Think of it like this: a variable made inside a function is like a person inside a room. People outside the room cannot see them.

Python looks for a variable name in this order, called the **LEGB rule**:
1. **L**ocal: Inside the current function
2. **E**nclosing: Inside the outer function (if there is a function inside a function)
3. **G**lobal: In the main program
4. **B**uilt-in: Python's own names like `print` and `len`

```python
city = "Kolkata"          # Global variable

def show():
    city = "Siliguri"      # Local variable (a different variable)
    print("Inside:", city)

show()
print("Outside:", city)
```

```python
def f():
    local_var = 10

f()
# print(local_var)        # This will raise a NameError (it only exists inside f)
```

## **The global Keyword**
We can **read** a global variable inside a function. But to **change** it, we must write `global`.

```python
counter = 0

def visit():
    global counter
    counter += 1

visit()
visit()
visit()
print(counter)
```

It is usually better to avoid `global`. A cleaner way is to pass the value in and return the new value.

```python
def visit(counter):
    return counter + 1

c = 0
c = visit(c)
c = visit(c)
print(c)
```

## **Enclosing Scope and nonlocal**
A function can be written inside another function. The inner function can read the outer function's variables. To **change** them, we use `nonlocal`.

```python
def trip_counter():
    trips = 0
    def add():
        nonlocal trips
        trips += 1
    add()
    add()
    return trips

print(trip_counter())
```

## **Don't Use Built-in Names as Variables**
If we use a name like `list` or `sum` for our own variable, the built-in function stops working.

```python
# list = [1, 2, 3]
# list("abc")        # This will raise a TypeError, because 'list' is now our variable

print(list("abc"))   # Works correctly because we did not replace 'list'
```

## **Returning Many Values**
A function can return many values at once. Python packs them into a **tuple**.

```python
def stats(nums):
    mean = sum(nums) / len(nums)
    return min(nums), max(nums), round(mean, 2)

aqi = [182, 210, 165, 240, 198, 150, 170]     # Kolkata winter week (sample data)
low, high, avg = stats(aqi)
print(f"Min {low}, Max {high}, Average {avg}")
```

## **Example from GIS: Converting Degrees, Minutes, Seconds**
Coordinates are often written in **DMS** format: **Degrees, Minutes, Seconds**. For example, Kolkata is at **22° 34' 21" N**.

GIS software needs **Decimal Degrees (DD)**, like **22.5725**.

**Formula:** DD = Degrees + Minutes / 60 + Seconds / 3600

(1 degree = 60 minutes, and 1 minute = 60 seconds, just like time.)

```python
def dms_to_dd(degrees, minutes, seconds, direction="N"):
    """Convert Degrees-Minutes-Seconds to Decimal Degrees."""
    dd = degrees + minutes / 60 + seconds / 3600
    if direction in ("S", "W"):
        dd = -dd                  # South and West are negative
    return round(dd, 6)

def dd_to_dms(dd):
    """Convert Decimal Degrees back to Degrees, Minutes, Seconds."""
    degrees = int(dd)
    minutes_float = (dd - degrees) * 60
    minutes = int(minutes_float)
    seconds = round((minutes_float - minutes) * 60, 2)
    return degrees, minutes, seconds

lat = dms_to_dd(22, 34, 21, "N")
lon = dms_to_dd(88, 21, 50, "E")
print("Kolkata in decimal degrees:", lat, lon)

d, m, s = dd_to_dms(26.7271)
print(f"Siliguri latitude in DMS: {d}° {m}' {s}\"")
```

Let us understand the code above:
- `minutes / 60`: 34 minutes = 34/60 = 0.5667 degree.
- `seconds / 3600`: 21 seconds = 21/3600 = 0.0058 degree.
- `dd_to_dms` returns **three values** as a tuple, which we unpack into `d, m, s`.

## **Returning a Function (a Preview)**
A function can also return another function. We will learn more about this on Day 64 (Closures).

```python
def make_multiplier(n):
    def multiply(x):
        return x * n
    return multiply

double = make_multiplier(2)
triple = make_multiplier(3)
print(double(10), triple(10))
```

## **Practice Problem: GST Calculator**
Return a dictionary with all the details.

```python
def gst(amount, rate=18):
    tax = round(amount * rate / 100, 2)
    return {"base": amount, "cgst": tax / 2, "sgst": tax / 2, "total": amount + tax}

print(gst(1000))
print(gst(500, rate=5))
```

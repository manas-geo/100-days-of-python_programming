# **Day 30: Mini Project - Madhyamik Result Analyzer**

In this project we use functions, loops, dictionaries and comprehensions together.

**The task:** A school in Midnapore has the Madhyamik (Class 10) marks of 10 students in 7 subjects. We want to:
1. Find the total, percentage and grade of each student.
2. Find the topper of each subject.
3. Find who failed and in which subject.
4. Make a summary for the whole class.

(All names and marks are made up.)

```python
subjects = ["bengali", "english", "maths", "phy_sc", "life_sc", "history", "geography"]

raw_data = """
Ankita,88,93,95,90,92,85,87
Soumya,72,65,58,61,70,66,69
Rupam,45,38,22,30,41,50,47
Tithi,94,90,98,96,95,92,93
Arnab,60,55,71,65,62,58,60
Moumita,81,77,69,74,80,79,95
Debjit,33,29,40,24,31,35,38
Priyanka,76,82,88,79,85,94,71
Sayan,55,48,99,87,62,44,50
Rimpa,69,71,63,58,66,72,70
"""
```

## **Step 1: Reading the Data**

```python
def parse(text):
    students = {}
    for line in text.strip().splitlines():
        name, *marks = line.split(",")
        students[name] = dict(zip(subjects, map(int, marks)))
    return students

students = parse(raw_data)
print(students["Tithi"])
```

- `name, *marks = line.split(",")`: The first value goes to `name`, and all the rest go into the list `marks`.
- `dict(zip(subjects, ...))`: Joins each subject name with its marks.

## **Step 2: Report for Each Student**

```python
PASS_MARK = 25

def grade(percent):
    for cutoff, g in [(90, "AA"), (80, "A+"), (60, "A"), (45, "B+"), (35, "B"), (25, "C")]:
        if percent >= cutoff:
            return g
    return "D"

def report(name, marks):
    total = sum(marks.values())
    percent = total / (len(marks) * 100) * 100
    failed = [s for s, m in marks.items() if m < PASS_MARK]
    return {"name": name, "total": total, "percent": round(percent, 2),
            "grade": grade(percent), "failed": failed}

reports = [report(n, m) for n, m in students.items()]
reports.sort(key=lambda r: r["total"], reverse=True)

print(f"{'Rank':<5}{'Name':<10}{'Total':>6}{'%':>8}{'Grade':>7}  Status")
print("-" * 48)
for i, r in enumerate(reports, 1):
    status = "PASS" if not r["failed"] else "Fail in " + ", ".join(r["failed"])
    print(f"{i:<5}{r['name']:<10}{r['total']:>6}{r['percent']:>8}{r['grade']:>7}  {status}")
```

## **Step 3: Subject Toppers**

```python
for s in subjects:
    best = max(students, key=lambda n: students[n][s])
    print(f"{s:<10} {best:<10} {students[best][s]}")
```

## **Step 4: Subject Averages as a Text Bar Chart**

```python
averages = {s: round(sum(st[s] for st in students.values()) / len(students), 1) for s in subjects}

for s, a in sorted(averages.items(), key=lambda kv: kv[1]):
    print(f"{s:<10} {a:>5} {'#' * int(a // 4)}")
```

## **Step 5: Class Summary**

```python
passed = [r for r in reports if not r["failed"]]
print(f"Students : {len(reports)}")
print(f"Passed   : {len(passed)} ({len(passed) / len(reports):.0%})")
print(f"Class average: {sum(r['percent'] for r in reports) / len(reports):.2f}%")

distribution = {g: sum(1 for r in reports if r["grade"] == g) for g in ["AA", "A+", "A", "B+", "B", "C", "D"]}
print("Grade distribution:", distribution)
```

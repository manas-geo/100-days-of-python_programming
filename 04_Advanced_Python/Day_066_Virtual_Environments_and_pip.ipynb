## **The Problem**
Suppose Project A needs version 1.5 of a library, and Project B needs version 2.2 of the same library. If both projects use the same Python installation, one of them will break.

## **The Solution: A Virtual Environment**
A **virtual environment** (venv) is a **separate, private box** of Python and libraries for **each project**. What we install in one box does not affect the others.

**Real-life example:** Each student has their own **school bag** with their own books. One student's books do not get mixed with another's.

## **Commands (Type These in the Terminal, Not in the Notebook)**

**1. Create a virtual environment**
```bash
python -m venv .venv
```

**2. Activate it**
```bash
# Windows
.venv\Scripts\activate

# Linux / Mac
source .venv/bin/activate
```
After activation, the terminal shows `(.venv)` at the start of the line.

**3. Install libraries with pip**
```bash
pip install requests
pip install pandas==2.2.2        # A fixed version
pip install --upgrade pip        # Update pip itself
pip uninstall requests
```

**4. Save the list of libraries, and install them again on another computer**
```bash
pip freeze > requirements.txt
pip install -r requirements.txt
```

**5. Leave the environment**
```bash
deactivate
```

## **Checking Things from Python**

````python
import sys
print(sys.version_info)
print("Inside a virtual environment:", sys.prefix != sys.base_prefix)
````

````python
from importlib.metadata import version, PackageNotFoundError

for package in ["pip", "numpy", "pandas", "requests", "some_fake_package"]:
    try:
        print(f"{package:<18} {version(package)}")
    except PackageNotFoundError:
        print(f"{package:<18} not installed")
````

## **Special Note for GIS Students: Installing GIS Libraries**
GIS libraries like **GDAL**, **Rasterio**, **GeoPandas** and **Fiona** depend on large programs written in C. Installing them with `pip` on Windows sometimes fails.

Two easy options:

**Option 1: conda (recommended for GIS)**
```bash
conda create -n gis python=3.12
conda activate gis
conda install -c conda-forge geopandas rasterio
```

**Option 2: pip (works on most computers now)**
```bash
pip install geopandas rasterio shapely pyproj
```

**Option 3: QGIS's own Python**

QGIS comes with its own Python, which already has GDAL. We can use it from the **Python Console** inside QGIS (Plugins -> Python Console).

## **Do Not Upload the venv Folder to GitHub**

